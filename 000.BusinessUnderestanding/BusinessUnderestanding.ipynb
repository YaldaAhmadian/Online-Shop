## سؤال ۱: مسئله اصلی کسب‌وکار چیست؟

مسئله اصلی فروشگاه آنلاین را با زبان ساده و از دیدگاه کسب‌وکار توضیح دهید.

مشخص کنید چه مشکلی در مدیریت رفتار مشتریان وجود دارد، چرا این مشکل برای فروشگاه اهمیت دارد و ادامه وضعیت فعلی چه پیامدهای مالی و عملیاتی خواهد داشت.

## چرا این سؤال مطرح می‌شود؟

این سؤال نقطه شروع مرحله Business Understanding است؛ زیرا پیش از انتخاب داده، الگوریتم یا روش تحلیلی باید مشخص کنیم سازمان واقعاً با چه مشکلی روبه‌رو است.

### علت اهمیت سؤال در این پروژه

در سناریوی Online Shop، مدیریت متوجه شده است که برخی مشتریان قدیمی به‌تدریج خرید خود را کاهش می‌دهند و برخی دیگر خرید را متوقف می‌کنند.

اما کاهش خرید مشتری صرفاً یک اتفاق آماری نیست؛ ممکن است به کاهش درآمدهای آینده، هدررفت منابع بازاریابی و از دست رفتن فرصت تعامل مجدد با مشتری منجر شود.

بنابراین باید بفهمیم مشکل اصلی سازمان چیست و چرا حل آن ارزش تجاری دارد.

- درک پیامدهای ادامه وضعیت فعلی

## پاسخ سؤال ۱: مسئله اصلی کسب‌وکار

مسئله اصلی این پروژه، ناتوانی فروشگاه آنلاین در شناسایی به‌موقع تغییرات رفتار خرید مشتریان و تشخیص افرادی است که احتمال دارد فعالیت خرید خود را متوقف کنند.

مشتریان فروشگاه از نظر تعداد سفارش‌ها، میزان هزینه‌کرد، فاصله زمانی بین خریدها، تنوع محصولات و سابقه خرید رفتارهای متفاوتی دارند.

برخی مشتریان خریدهای منظم انجام می‌دهند، برخی به‌صورت مقطعی خرید می‌کنند و برخی دیگر ممکن است به‌تدریج تعداد سفارش‌ها یا میزان خرید خود را کاهش دهند.

مشکل زمانی ایجاد می‌شود که فروشگاه نتواند این تغییرات را به‌موقع تشخیص دهد و مشتریان مناسب را برای اقدامات بازاریابی انتخاب کند.

### پیامدهای ادامه وضعیت فعلی

۱. **از دست رفتن درآمد بالقوه:** توقف خرید مشتریان می‌تواند فرصت فروش‌های تکراری و درآمد آینده فروشگاه را کاهش دهد.

۲. **کاهش اثربخشی بازاریابی:** اجرای کمپین‌های عمومی بدون شناخت وضعیت مشتریان می‌تواند اثربخشی منابع بازاریابی را کاهش دهد.

۳. **هدررفت بودجه بازاریابی:** ممکن است منابع برای مشتریانی هزینه شوند که نیازی به مداخله ندارند یا احتمال پاسخ‌گویی آن‌ها پایین است.

۴. **عدم شناسایی تغییر رفتار:** فروشگاه نمی‌تواند به‌صورت نظام‌مند تغییراتی مانند افزایش فاصله بین خریدها یا کاهش تعداد سفارش‌ها را تحلیل کند.

۵. **از دست رفتن فرصت حفظ مشتریان ارزشمند:** مشتریان دارای سابقه خرید قابل توجه ممکن است بدون شناسایی قبلی از چرخه خرید خارج شوند.

۶. **ضعف در تصمیم‌گیری مدیریتی:** مدیران اطلاعات کافی برای اولویت‌بندی مشتریان و تخصیص هدفمند منابع نخواهند داشت.

### تعریف نهایی مسئله کسب‌وکار

> ناتوانی فروشگاه آنلاین در شناسایی زودهنگام مشتریان در معرض توقف خرید، شناخت الگوهای رفتاری آن‌ها و تخصیص هدفمند منابع بازاریابی برای حفظ مشتریان و کاهش فرصت‌های درآمدی ازدست‌رفته.

بنابراین مسئله اصلی، صرفاً پیش‌بینی یک برچسب نیست؛ بلکه بهبود توانایی فروشگاه در شناخت مشتریان و تصمیم‌گیری درباره نحوه حفظ آن‌هاست.

## سؤال ۲: ذی‌نفعان (Stakeholders) این پروژه چه کسانی هستند و هرکدام چه نیازها و انتظاراتی دارند؟

- موفقیت پروژه را از چه دیدگاهی ارزیابی می‌کند؟

### تفاوت نیاز و انتظار

نیاز (Need) مشکلی است که ذی‌نفع می‌خواهد برطرف شود.

انتظار (Expectation) نتیجه‌ای است که از اجرای پروژه امیدوار است به دست آورد.

مثال:

- نیاز مدیر بازاریابی: شناسایی مشتریانی که احتمال توقف خرید آن‌ها بالاست.
- انتظار مدیر بازاریابی: دریافت فهرست مشتریان همراه با امتیاز ریسک و اطلاعات رفتاری قابل استفاده.

### تعارض‌های احتمالی

- مدیریت بازاریابی ممکن است به دنبال افزایش تعداد تماس‌ها باشد.
- واحد مالی ممکن است کاهش هزینه کمپین را در اولویت قرار دهد.
- تیم CRM ممکن است به اطلاعات رفتاری بیشتر نیاز داشته باشد.
- واحد حریم خصوصی ممکن است دسترسی به بعضی اطلاعات را محدود کند.

بنابراین باید میان ارزش تجاری، هزینه اجرا، کاربردپذیری و محرمانگی تعادل ایجاد شود.

### جمع‌بندی

موفقیت پروژه زمانی امکان‌پذیر است که خروجی‌های تحلیلی علاوه بر اعتبار فنی، متناسب با نیازهای تصمیم‌گیرندگان و کاربران سازمان باشند.

## پاسخ سؤال ۳: Business Objectives

### هدف اصلی کسب‌وکار

افزایش توانایی فروشگاه آنلاین در حفظ مشتریان و کاهش فرصت‌های درآمدی ازدست‌رفته، از طریق تحلیل رفتار خرید، شناسایی مشتریان در معرض توقف خرید و طراحی راهبردهای هدفمند بازاریابی.

### Business Objective Statement

> ایجاد یک چارچوب تصمیم‌یار مبتنی بر داده برای تحلیل رفتار مشتریان، شناسایی افراد در معرض توقف خرید و اولویت‌بندی اقدامات حفظ مشتری با هدف بهبود اثربخشی کمپین‌های بازاریابی و استفاده بهینه از منابع سازمان.

### اهداف فرعی (Sub Objectives)

**۱. شناخت ساختار رفتاری مشتریان**

شناسایی گروه‌های مختلف مشتریان با استفاده از تحلیل ویژگی‌های رفتاری و چارچوب LRRFM به‌منظور شناخت تفاوت مشتریان از نظر ابعاد مختلف رابطه و رفتار خرید.

**۲. شناسایی مشتریان در معرض توقف خرید**

پیش‌بینی مشتریانی که طبق تعریف عملیاتی پروژه، در ۹۰ روز آینده خرید معتبر جدیدی نخواهند داشت.

**۳. تحلیل عوامل مرتبط با توقف خرید**

بررسی ارتباط ویژگی‌های رفتاری، شاخص‌های LRRFM و تغییرات رفتار خرید با احتمال عدم خرید آینده.

**۴. طراحی راهبردهای متفاوت حفظ مشتری**

ارائه پیشنهادهای بازاریابی متناسب با پروفایل رفتاری و سطح ریسک مشتریان.

**۵. اولویت‌بندی مشتریان برای کمپین**

انتخاب ۵۰۰ مشتری بر اساس ریسک پیش‌بینی‌شده و معیارهای تجاری مورد توافق.

**۶. ارزیابی اقتصادی کمپین**

برآورد هزینه، سود افزایشی، نقطه سربه‌سر و حساسیت نتایج نسبت به فرضیات اقتصادی.

### ارتباط اهداف فرعی

LRRFM به شناخت چندبعدی رفتار و رابطه مشتری با فروشگاه کمک می‌کند. این شناخت در کنار سایر ویژگی‌های رفتاری، مبنای خوشه‌بندی اولیه و تحلیل‌های بعدی خواهد بود.

درصد مشتریانی که در یک دوره مشخص همچنان به خرید از فروشگاه ادامه می‌دهند.

این شاخص برای ارزیابی توانایی فروشگاه در حفظ مشتری اهمیت دارد.

### ۲. نرخ بازگشت مشتریان منتخب

درصد مشتریان هدف کمپین که پس از اجرای مداخله دوباره خرید انجام می‌دهند.

برای سنجش اثر واقعی کمپین باید این نرخ با گروه کنترل مقایسه شود.

### ۳. سود افزایشی کمپین (Incremental Profit)

سود اضافی ایجادشده در نتیجه اجرای کمپین نسبت به شرایطی که مداخله انجام نشده است.

این شاخص باید با طراحی آزمایش مناسب اندازه‌گیری شود.

### ۴. بازگشت سرمایه کمپین (Campaign ROI)

میزان بازده اقتصادی کمپین نسبت به هزینه آن.

ROI = (Incremental Profit - Campaign Cost) / Campaign Cost × 100

### ۵. هزینه حفظ هر مشتری (Cost per Incremental Retained Customer)

هزینه کمپین تقسیم بر تعداد مشتریانی که به‌صورت افزایشی در اثر مداخله حفظ شده‌اند.

### ۶. بهره‌وری ظرفیت تماس

بررسی میزان استفاده مؤثر از ظرفیت تماس با ۵۰۰ مشتری.

### نکته: معیارهای انتخاب مشتری

Precision@500، Recall@500 و Lift@500 نیز برای سنجش کیفیت فهرست انتخاب‌شده مفیدند، اما این‌ها معیارهای ارزیابی رتبه‌بندی مدل هستند و به‌تنهایی Business KPI محسوب نمی‌شوند.

### نحوه تعیین Target

در سناریو مقدار پایه، هدف عددی و داده واقعی کمپین ارائه نشده است.

بنابراین ابتدا باید Baseline از داده و اطلاعات تجاری تعیین شود و سپس هدف واقع‌بینانه برای هر شاخص مشخص گردد.

### جمع‌بندی

## پاسخ سؤال ۷: Project Constraints & Assumptions

### الف) محدودیت‌های داده‌ای

**۴. تراکنش‌های غیرعادی**

اطلاعات سود خالص، هزینه کمپین، سوابق تخفیف و رضایت مشتری در ستون‌های معرفی‌شده وجود ندارند.

| H5 | گروه‌های رفتاری مختلف ممکن است به راهبردهای متفاوت نیاز داشته باشند. | تحلیل خوشه‌ها و آزمون کمپین |
| H6 | اولویت‌بندی مشتریان می‌تواند استفاده از ظرفیت ۵۰۰ تماس را بهبود دهد. | مقایسه عملکرد انتخاب با روش مبنا |

این موارد فرضیه هستند و نباید پیش از تحلیل داده به‌عنوان یافته قطعی گزارش شوند.

## پاسخ سؤال ۱۰: SWOT Analysis

### Strengths ـ نقاط قوت

- امکان تحلیل الگوی خرید در سطح مشتری

### Weaknesses ـ نقاط ضعف

- محدود بودن بازه تاریخی
- وجود CustomerID گمشده
- وجود تراکنش‌های منفی و لغوشده

- احتمال ناکارآمدی مداخله بازاریابی
- محدودیت‌های قانونی مربوط به داده مشتریان

### راهبردهای حاصل از SWOT

**استفاده از قوت‌ها برای فرصت‌ها (SO)**

استفاده از داده‌های تراکنشی و زمانی برای کشف الگوهای رفتاری و شناسایی مشتریان دارای رفتار مشابه.

**استفاده از قوت‌ها برای کاهش تهدیدها (ST)**

استفاده از تاریخ‌های مرجع مختلف و ارزیابی زمانی برای بررسی پایداری نتایج.

**کاهش ضعف‌ها برای استفاده از فرصت‌ها (WO)**

مستندسازی و اصلاح مشکلات داده و ساخت ویژگی‌های معتبر در سطح مشتری.

**کاهش ضعف‌ها برای مقابله با تهدیدها (WT)**

محدود کردن ادعاهای مدل به تعریف عملیاتی پروژه و استفاده از آزمایش کنترل‌شده برای ارزیابی اثر کمپین.

### جمع‌بندی

SWOT نشان می‌دهد پروژه ظرفیت مناسبی برای تحلیل رفتار مشتری دارد، اما کیفیت داده، محدودیت زمانی و نبود اطلاعات اقتصادی باید در تصمیم‌های بعدی لحاظ شوند.

## پاسخ سؤال ۱۱: Business Goal vs Data Mining Goal

### هدف کسب‌وکار (Business Goal)

افزایش توانایی فروشگاه آنلاین در حفظ مشتریان، کاهش فرصت‌های درآمدی ازدست‌رفته و بهبود اثربخشی فعالیت‌های بازاریابی.

### هدف داده‌کاوی (Data Mining Goal)

تولید اطلاعات تحلیلی و پیش‌بینی‌های معتبر از رفتار مشتریان در سه بخش اصلی:

**۱. Customer Segmentation**

کشف گروه‌های رفتاری مشتریان با استفاده از ویژگی‌های رفتاری و چارچوب LRRFM.

هدف این مرحله شناسایی مشتریان دارای الگوهای خرید و ویژگی‌های رابطه‌ای مشابه است.

**۲. Churn Risk Prediction**

ساخت مدل طبقه‌بندی برای پیش‌بینی عدم خرید معتبر مشتری در ۹۰ روز آینده، بر اساس ویژگی‌های استخراج‌شده از ۱۸۰ روز گذشته.

ویژگی‌های LRRFM می‌توانند در کنار سایر ویژگی‌های رفتاری، به‌عنوان بخشی از ورودی این مدل استفاده شوند.

**۳. Advanced Customer Segmentation**

کشف گروه‌های تکمیلی مشتریان با استفاده از پروفایل رفتاری و امتیاز ریسک پیش‌بینی‌شده.

هدف، ایجاد گروه‌هایی قابل استفاده برای طراحی راهبردهای متفاوت حفظ مشتری است.

### Data Mining Goal Statement

> تحلیل داده‌های تراکنشی مشتریان با استفاده از ویژگی‌های رفتاری مبتنی بر LRRFM و سایر شاخص‌های خرید، به‌کارگیری روش‌های خوشه‌بندی برای کشف گروه‌های رفتاری، توسعه مدل طبقه‌بندی برای پیش‌بینی عدم خرید معتبر در افق ۹۰ روزه و ایجاد گروه‌های تحلیلی تکمیلی برای پشتیبانی از تصمیم‌گیری‌های حفظ مشتری.

## پاسخ سؤال ۱۲: Data Mining Task Identification

معماری تحلیلی پروژه شامل سه مرحله اصلی است:

Clustering → Classification → Clustering

### مرحله اول: Clustering

**نوع یادگیری:** Unsupervised Learning

**هدف:**

کشف گروه‌های طبیعی مشتریان بر اساس شباهت ویژگی‌های رفتاری آن‌ها.

**ورودی:**

ویژگی‌های رفتاری مشتریان شامل:

- ویژگی‌های مبتنی بر LRRFM
- تعداد سفارش‌ها
- میزان هزینه‌کرد
- تنوع محصولات
- تعداد روزهای خرید
- الگوهای زمانی خرید

ویژگی‌های LRRFM در این مرحله نقش مهمی در توصیف ابعاد مختلف رفتار مشتری خواهند داشت.

**خروجی:**

Cluster Label و پروفایل رفتاری هر خوشه.

**کاربرد:**

شناخت ساختار مشتریان و کشف تفاوت گروه‌های رفتاری.

### مرحله دوم: Classification

**نوع یادگیری:** Supervised Learning

**هدف:**

پیش‌بینی عدم خرید معتبر مشتری در ۹۰ روز آینده.

**برچسب هدف:**

- Churn = 1: هیچ خرید معتبر در ۹۰ روز آینده وجود ندارد.
- Churn = 0: حداقل یک خرید معتبر در ۹۰ روز آینده وجود دارد.

**ورودی:**

ویژگی‌های رفتاری محاسبه‌شده از ۱۸۰ روز پیش از تاریخ مرجع، شامل ویژگی‌های معتبر LRRFM و سایر ویژگی‌های مهندسی‌شده.

**خروجی:**

کلاس پیش‌بینی‌شده و امتیاز یا احتمال ریسک عدم خرید.

**کاربرد:**

شناسایی و اولویت‌بندی مشتریان با ریسک بالاتر.

### مرحله سوم: Clustering تکمیلی

**نوع یادگیری:** Unsupervised Learning

**هدف:**

کشف گروه‌های تکمیلی مشتریان با توجه به ویژگی‌های رفتاری و ریسک پیش‌بینی‌شده.

**ورودی‌های احتمالی:**

- ویژگی‌های معتبر LRRFM
- سایر ویژگی‌های رفتاری
- پروفایل خوشه‌بندی اولیه
- امتیاز ریسک Classification که به‌صورت خارج از نمونه تولید شده باشد.

**خروجی:**

گروه‌های تحلیلی مناسب برای طراحی راهبردهای متفاوت حفظ مشتری.

### مقایسه سه مرحله

| مرحله | روش | ورودی اصلی | خروجی |
|---|---|---|---|
| اول | Clustering | LRRFM و رفتار خرید | گروه‌های رفتاری |
| دوم | Classification | ویژگی‌های رفتاری گذشته | امتیاز ریسک و کلاس Churn |
| سوم | Clustering | پروفایل رفتاری و ریسک پیش‌بینی‌شده | گروه‌های عملیاتی تکمیلی |

### نکته مهم

در هیچ‌یک از مراحل خوشه‌بندی، برچسب واقعی Churn نباید به‌عنوان ویژگی ورودی استفاده شود.

همچنین استفاده از Cluster Label مرحله اول در Classification باید به‌صورت آزمایشی ارزیابی شود تا مشخص شود آیا واقعاً ارزش پیش‌بینی اضافه ایجاد می‌کند یا خیر.

تعریف دقیق پنج بُعد LRRFM نیز باید پیش از محاسبه و ورود به مدل‌سازی نهایی شود.

### جمع‌بندی

LRRFM بخشی از طراحی ویژگی‌های رفتاری پروژه است و در خوشه‌بندی اولیه و در صورت اعتبارسنجی، طبقه‌بندی و خوشه‌بندی تکمیلی نیز کاربرد خواهد داشت.

## پاسخ سؤال ۱۳: Data Mining Evaluation Criteria

معیارهای ارزیابی باید متناسب با ماهیت هر تسک انتخاب شوند.

### الف) ارزیابی خوشه‌بندی اولیه

از آنجا که هدف این مرحله کشف گروه‌های رفتاری است، باید هم کیفیت ریاضی خوشه‌ها و هم معناداری رفتاری آن‌ها بررسی شود.

معیارهای پیشنهادی:

**۱. Silhouette Score**

سنجش میزان نزدیکی نمونه به خوشه خودش و فاصله آن از خوشه‌های دیگر.

**۲. Davies-Bouldin Index**

ارزیابی پراکندگی درون خوشه‌ها و شباهت بین خوشه‌ها.

**۳. Calinski-Harabasz Index**

بررسی نسبت پراکندگی بین خوشه‌ها به پراکندگی درون خوشه‌ها.

**۴. Cluster Stability**

بررسی اینکه آیا ساختار خوشه‌ها با تغییرات جزئی داده یا نمونه‌گیری همچنان پایدار باقی می‌ماند.

**۵. Behavioral Interpretability**

بررسی اینکه آیا خوشه‌ها از نظر ویژگی‌های رفتاری و LRRFM تفاوت قابل توضیحی دارند یا خیر.

در این بخش باید بررسی شود آیا مشتریان هر خوشه از نظر ابعاد LRRFM واقعاً رفتار متفاوتی دارند.

### ب) ارزیابی Classification

معیارهای پیشنهادی:

- Precision
- Recall
- F1-score
- PR-AUC
- ROC-AUC
- Confusion Matrix

Accuracy نیز گزارش می‌شود، اما به‌تنهایی برای انتخاب مدل کافی نیست.

### ج) تحلیل False Positive و False Negative

**False Positive:**

مشتری به‌عنوان در معرض ریزش پیش‌بینی شده، اما در ۹۰ روز آینده خرید معتبر داشته است.

هزینه احتمالی آن، مصرف ظرفیت تماس و ارائه پیشنهاد غیرضروری است.

**False Negative:**

مشتری طبق تعریف پروژه در آینده خریدی نداشته، اما مدل او را شناسایی نکرده است.

این موضوع می‌تواند باعث ازدست‌رفتن فرصت تعامل مجدد با مشتری شود.

### د) ارزیابی زمانی

- ویژگی‌ها فقط از ۱۸۰ روز گذشته ساخته شوند.
- برچسب فقط با بررسی ۹۰ روز آینده تعیین شود.
- مدل فقط از برچسب‌های قابل دسترس در زمان آموزش استفاده کند.
- تمام پیش‌پردازش و انتخاب ویژگی در چارچوب ارزیابی مناسب انجام شود.
- هم‌پوشانی پنجره‌ها و وابستگی نمونه‌های یک مشتری کنترل شود.

### هـ) ارزیابی خوشه‌بندی تکمیلی

علاوه بر معیارهای عددی خوشه‌بندی، باید بررسی کنیم:

- آیا گروه‌های جدید با خوشه‌بندی اولیه تفاوت دارند؟
- آیا پروفایل LRRFM گروه‌ها متفاوت است؟
- آیا سطح ریسک پیش‌بینی‌شده بین گروه‌ها تفاوت دارد؟
- آیا این گروه‌ها برای طراحی کمپین کاربرد دارند؟
- آیا ترکیب ویژگی‌های رفتاری و ریسک، ارزش افزوده‌ای نسبت به روش پایه ایجاد کرده است؟

### جمع‌بندی

LRRFM در این پروژه یکی از مبانی تفسیر رفتار و ارزیابی معناداری خوشه‌ها خواهد بود.

اما کیفیت خوشه‌بندی فقط با نام یا تعداد ابعاد LRRFM مشخص نمی‌شود؛ باید پایداری، جدایی و کاربرد تجاری گروه‌ها نیز بررسی شود.

```text
Online Shop Data Mining Project
│
├── 1. Business Understanding
│   ├── Business Problem
│   ├── Business Objectives
│   ├── Stakeholder Analysis
│   ├── Constraints & Risks
│   └── Data Mining Goals
│
├── 2. Data Understanding
│   ├── Dataset Acquisition
│   ├── Data Structure Analysis
│   ├── Data Quality Assessment
│   ├── Statistical Analysis
│   └── EDA
│
├── 3. Data Preparation
│   ├── Transaction Validation
│   ├── Missing Values
│   ├── Duplicate Records
│   ├── Cancellation & Return Analysis
│   ├── Customer Aggregation
│   ├── LRRFM Feature Engineering
│   ├── Behavioral Feature Engineering
│   └── Churn Label Construction
│
├── 4. Modeling
│   ├── Initial Clustering
│   ├── Cluster Profiling
│   ├── Classification
│   ├── Model Comparison
│   └── Advanced Clustering
│
├── 5. Evaluation
│   ├── Clustering Evaluation
│   ├── LRRFM-based Cluster Analysis
│   ├── Classification Evaluation
│   ├── Temporal Validation
│   ├── Leakage Assessment
│   └── Feature Interpretation
│
└── 6. Deployment
    ├── Customer Ranking
    ├── Campaign Selection
    ├── Campaign Strategy
    ├── Economic Analysis
    └── Management Reporting

**سلول ۱ ـ سؤال**

<div dir="rtl" align="right">

## سؤال ۱۶: برای اجرای پروژه چه زمان‌بندی و وابستگی‌هایی باید در نظر گرفته شود؟

برای هر فاز CRISP-DM برنامه زمانی پیشنهادی تهیه کنید.

مشخص کنید:

- چه فعالیت‌هایی باید پیش از فعالیت‌های دیگر انجام شوند؟
- چه فعالیت‌هایی می‌توانند موازی اجرا شوند؟
- مهم‌ترین Milestoneهای پروژه کدام‌اند؟
- تأخیر در هر مرحله چه تأثیری بر مراحل بعدی دارد؟
- چگونه می‌توان زمان‌بندی را در صورت بروز تأخیر اصلاح کرد؟

</div>

<div dir="rtl" align="right">

## چرا این سؤال مطرح می‌شود؟

پروژه‌های داده‌کاوی فعالیت‌های متعددی دارند و برخی فعالیت‌ها برای شروع به خروجی فعالیت‌های قبلی وابسته‌اند.

بدون زمان‌بندی و مشخص کردن وابستگی‌ها، کنترل پیشرفت پروژه دشوار می‌شود.

### علت اهمیت سؤال در این پروژه

برای مثال، مدل‌سازی به داده آماده و ویژگی‌های معتبر نیاز دارد و ساخت برچسب به تعریف درست پنجره‌های زمانی وابسته است.

بنابراین نمی‌توان تمام مراحل را مستقل و بدون ترتیب اجرا کرد.

### هدف پرسش

- تعیین توالی منطقی فعالیت‌ها
- تخمین زمان اجرای هر بخش
- شناسایی فعالیت‌های قابل اجرای موازی
- تعیین نقاط کنترل پروژه
- مدیریت اثر تأخیرها

### فایده پاسخ

به جلوگیری از دوباره‌کاری، تأخیرهای زنجیره‌ای و حذف عجولانه فعالیت‌های ضروری کمک می‌کند.

### خروجی مورد انتظار

Project Timeline، Dependency Map و Milestone Plan.

</div>

<div dir="rtl" align="right">

## پاسخ سؤال ۱۶: Timeline & Dependencies

### الف) برنامه زمانی پیشنهادی

مدت دقیق هر مرحله باید بر اساس تعداد اعضای تیم، پیچیدگی اجرا و مهلت پروژه تعیین شود.

| فاز | فعالیت | وابستگی |
|---|---|---|
| Business Understanding | تعریف مسئله و برنامه | شروع پروژه |
| Data Understanding | دریافت و شناخت داده | دسترسی به دیتاست |
| Data Preparation | پاک‌سازی و مهندسی ویژگی | شناخت ساختار داده |
| Modeling | اجرای سه مرحله تحلیلی | آماده بودن داده و طراحی ویژگی‌ها |
| Evaluation | ارزیابی نتایج | تولید مدل و خروجی‌ها |
| Deployment | تهیه خروجی نهایی و پیشنهاد کمپین | تکمیل ارزیابی |

### ب) وابستگی‌های مهم

1. تعریف مسئله پیش‌نیاز تعیین هدف داده‌کاوی است.
2. شناخت داده پیش‌نیاز طراحی معتبر Data Preparation است.
3. قوانین اعتبارسنجی تراکنش باید قبل از محاسبه ویژگی‌ها مشخص شوند.
4. تعریف تاریخ مرجع و پنجره‌ها باید پیش از ساخت برچسب مشخص شود.
5. Modeling به داده و ویژگی‌های آماده نیاز دارد.
6. انتخاب خروجی نهایی به نتایج Evaluation وابسته است.

### ج) فعالیت‌های قابل اجرای موازی

- مستندسازی هم‌زمان با تحلیل داده
- بررسی داده‌های تکمیلی هم‌زمان با آماده‌سازی داده
- طراحی اولیه گزارش‌ها در کنار Modeling
- بررسی فرضیات اقتصادی هم‌زمان با تحلیل رفتاری

### د) Milestoneها

| کد | نقطه عطف | معیار تکمیل |
|---|---|---|
| M1 | پایان Business Understanding | تأیید اهداف و برنامه |
| M2 | پایان Data Understanding | تکمیل گزارش شناخت داده |
| M3 | پایان Data Preparation | تأیید قوانین داده و خروجی آماده |
| M4 | تکمیل Clustering اول | تحلیل و پروفایل خوشه‌ها |
| M5 | تکمیل Classification | آموزش و مقایسه مدل‌ها |
| M6 | تکمیل Clustering دوم | تحلیل خوشه‌های تکمیلی |
| M7 | پایان Evaluation | تأیید گزارش ارزیابی |
| M8 | پایان Deployment | تحویل خروجی‌های نهایی |

### هـ) مدیریت تأخیر

در صورت تأخیر در Data Preparation می‌توان فعالیت‌های مستقل مانند مستندسازی یا طراحی گزارش را ادامه داد.

اما نباید برای جبران تأخیر، اعتبارسنجی داده یا ارزیابی زمانی حذف شوند.

### جمع‌بندی

زمان‌بندی باید بر اساس وابستگی واقعی فعالیت‌ها طراحی شود، نه صرفاً تقسیم مساوی زمان بین شش فاز.

</div>

<div dir="rtl" align="right">

## سؤال ۱۷: چه شرایطی برای پذیرش نهایی پروژه لازم است و چه عواملی ممکن است باعث شکست آن شوند؟

معیارهای پذیرش را از سه دیدگاه بررسی کنید:

- کسب‌وکار
- فنی و تحلیلی
- مدیریتی و اجرایی

سپس عوامل شکست پروژه را شناسایی کنید و توضیح دهید چگونه می‌توان از آن‌ها جلوگیری کرد.

</div>

<div dir="rtl" align="right">

## چرا این سؤال مطرح می‌شود؟

تکمیل فعالیت‌های برنامه‌ریزی‌شده لزوماً به معنای موفقیت پروژه نیست.

باید از ابتدا مشخص کنیم چه شرایطی نشان می‌دهد خروجی نهایی قابل قبول است.

### علت اهمیت سؤال در این پروژه

ممکن است مدل عملکرد آماری مناسبی داشته باشد، اما فهرست ۵۰۰ مشتری برای تیم بازاریابی کاربردی نباشد یا تحلیل اقتصادی آن قابل دفاع نباشد.

بنابراین معیار پذیرش باید هم کیفیت فنی و هم ارزش تجاری را در نظر بگیرد.

### هدف پرسش

- تعریف شرایط پذیرش پروژه
- جلوگیری از ارزیابی سلیقه‌ای
- مشخص کردن حداقل خروجی قابل قبول
- شناسایی عوامل شکست
- ایجاد ارتباط بین کیفیت فنی و نیاز کسب‌وکار

### فایده پاسخ

به تصمیم‌گیری درباره تکمیل، اصلاح یا عدم پذیرش پروژه کمک می‌کند.

### خروجی مورد انتظار

Project Acceptance Criteria و فهرست Failure Factors.

</div>

<div dir="rtl" align="right">

## پاسخ سؤال ۱۷: Acceptance Criteria

### الف) معیارهای پذیرش کسب‌وکاری

- مسئله و اهداف اصلی پروژه به‌درستی پوشش داده شده باشند.
- نتایج تحلیل برای تصمیم‌گیری بازاریابی قابل استفاده باشند.
- فهرست ۵۰۰ مشتری طبق معیارهای انتخاب مشخص تولید شده باشد.
- برای گروه‌های مختلف مشتریان اقدامات پیشنهادی تعریف شده باشد.
- تحلیل اقتصادی همراه با فرضیات و محدودیت‌ها ارائه شده باشد.

### ب) معیارهای پذیرش فنی

- قوانین اعتبارسنجی تراکنش‌ها مستند باشند.
- فرایند ساخت ویژگی‌ها و برچسب قابل بازتولید باشد.
- سه مرحله Clustering، Classification و Clustering اجرا شده باشند.
- حداقل چهار مدل طبقه‌بندی و یک Baseline مقایسه شده باشند.
- ارزیابی زمانی انجام شده باشد.
- نشت اطلاعات بررسی و کنترل شده باشد.
- معیارهای ارزیابی متناسب با هر تسک گزارش شده باشند.

### ج) معیارهای پذیرش مدیریتی و اجرایی

- گزارش نهایی قابل فهم باشد.
- محدودیت‌ها و ریسک‌ها شفاف گزارش شده باشند.
- پیشنهادهای کمپین با ظرفیت ۵۰۰ تماس سازگار باشند.
- خروجی برای تیم بازاریابی قابل استفاده باشد.
- شناسه مشتریان در خروجی‌های ارائه‌شده مستعارسازی شده باشد.

### د) عوامل احتمالی شکست

۱. تعریف نادرست مسئله و برچسب ریزش.

۲. کیفیت نامناسب داده.

۳. نشت اطلاعات آینده.

۴. ارزیابی غیرواقعی مدل.

۵. خوشه‌بندی غیرقابل تفسیر.

۶. نبود داده اقتصادی معتبر.

۷. انتخاب نامناسب مشتریان برای کمپین.

۸. فرض کردن اینکه پیش‌بینی ریسک به معنای اثربخشی قطعی تخفیف است.

۹. عدم پذیرش خروجی توسط تیم بازاریابی.

۱۰. تعمیم نامناسب نتایج به دوره‌های زمانی جدید.

### جمع‌بندی

پروژه زمانی قابل پذیرش است که خروجی‌های آن از نظر فنی معتبر، از نظر مدیریتی قابل استفاده و از نظر کسب‌وکار مرتبط با مسئله اصلی باشند.

</div>

<div dir="rtl" align="right">

## سؤال ۱۸: یک خلاصه مدیریتی (Executive Summary) برای پروژه تحلیل رفتار خرید و پیش‌بینی ریزش مشتریان تهیه کنید.

در این خلاصه، موارد زیر را پوشش دهید:

- مسئله اصلی و اهمیت آن برای کسب‌وکار
- هدف اصلی پروژه
- داده‌های مورد استفاده
- تعریف عملیاتی ریزش مشتری
- معماری تحلیلی و وظایف داده‌کاوی
- نتایج و خروجی‌های مورد انتظار
- کاربرد نتایج در تصمیم‌گیری و کمپین حفظ مشتری
- ارزش اقتصادی مورد انتظار
- محدودیت‌ها و ریسک‌های اصلی

متن باید رسمی، منسجم و قابل درک برای مدیرانی باشد که لزوماً دانش تخصصی داده‌کاوی ندارند.

</div>

<div dir="rtl" align="right">

## چرا این سؤال مطرح می‌شود؟

در پروژه‌های حرفه‌ای، مدیران ارشد معمولاً فرصت بررسی تمام جزئیات فنی، کدنویسی و تحلیل‌های آماری را ندارند.

بنابراین باید بتوانیم مسئله، اهداف، راهبرد اجرایی، خروجی‌ها و ارزش تجاری پروژه را در قالب یک سند خلاصه و یکپارچه ارائه کنیم.

### علت اهمیت سؤال در این پروژه

پروژه Online Shop شامل چندین فعالیت مرتبط با تحلیل رفتار مشتری، خوشه‌بندی، پیش‌بینی ریسک عدم خرید، تحلیل عوامل مؤثر و طراحی کمپین حفظ مشتری است.

اگر ارتباط این فعالیت‌ها با هدف اصلی کسب‌وکار مشخص نباشد، ممکن است پروژه از نگاه مدیریت صرفاً مجموعه‌ای از تحلیل‌های فنی به نظر برسد.

### هدف پرسش

۱. خلاصه‌سازی مسئله و اهداف کسب‌وکار.

۲. نشان دادن ارتباط بین اهداف کسب‌وکار و اهداف داده‌کاوی.

۳. معرفی معماری تحلیلی پروژه.

۴. مشخص کردن خروجی‌های مورد انتظار.

۵. بیان ارزش تجاری و محدودیت‌های پروژه.

۶. فراهم کردن مبنایی برای تصمیم‌گیری مدیریتی.

### فایده پاسخ

این سند به مدیران کمک می‌کند بدون ورود به جزئیات الگوریتم‌ها و کدنویسی، ضرورت اجرای پروژه، مسیر پیشنهادی و نتایج مورد انتظار را درک کنند.

همچنین به‌عنوان خلاصه‌ای از سند Business Understanding، مبنای ورود به مراحل بعدی CRISP-DM خواهد بود.

### خروجی مورد انتظار

Executive Summary شامل تعریف مسئله، اهداف، راهبرد، ارزش مورد انتظار، محدودیت‌ها و نتایج قابل تحویل.

</div>

<div dir="rtl" align="right">

## پاسخ سؤال ۱۸: Executive Summary

### پروژه تحلیل رفتار خرید، پیش‌بینی ریزش و طراحی راهبرد حفظ مشتری

### ۱. شرح مسئله کسب‌وکار

یکی از چالش‌های مهم فروشگاه‌های آنلاین، شناسایی مشتریانی است که به‌تدریج فعالیت خرید خود را کاهش می‌دهند یا برای مدتی خرید از فروشگاه را متوقف می‌کنند.

عدم شناسایی به‌موقع این تغییرات رفتاری می‌تواند باعث از دست رفتن فرصت‌های درآمدی آینده، کاهش اثربخشی فعالیت‌های بازاریابی و تخصیص نامناسب منابع سازمان شود.

در سناریوی حاضر، مدیریت فروشگاه قصد دارد با استفاده از تحلیل داده‌های تراکنشی، مشتریان را بهتر بشناسد، افراد در معرض توقف خرید را شناسایی کند و برای حفظ آن‌ها راهبردهای مناسبی طراحی کند.

### ۲. هدف پروژه

هدف اصلی پروژه، توسعه یک چارچوب تحلیلی و تصمیم‌یار مبتنی بر داده برای شناخت رفتار مشتریان، پیش‌بینی عدم خرید آینده و پشتیبانی از تصمیم‌گیری‌های مرتبط با حفظ مشتری است.

این چارچوب باید علاوه بر شناسایی مشتریان در معرض ریسک، امکان تحلیل گروه‌های رفتاری و اولویت‌بندی مشتریان برای اقدامات بازاریابی را فراهم کند.

### ۳. داده‌های مورد استفاده

منبع داده پروژه، دیتاست UCI Online Retail است.

این دیتاست شامل اطلاعات تراکنش‌های فروشگاه آنلاین در بازه زمانی دسامبر ۲۰۱۰ تا دسامبر ۲۰۱۱ است.

ستون‌های اصلی شامل شماره فاکتور، کد محصول، توضیحات محصول، تعداد کالا، تاریخ تراکنش، قیمت واحد، شناسه مشتری و کشور مشتری هستند.

پیش از استفاده از داده‌ها، کیفیت و اعتبار تراکنش‌ها از نظر مقادیر گمشده، رکوردهای تکراری، فاکتورهای لغوشده، مرجوعی‌ها و مقادیر غیرعادی بررسی خواهد شد.

### ۴. معماری تحلیلی پروژه

راهبرد داده‌کاوی پروژه در سه مرحله اصلی طراحی شده است:

**مرحله اول: خوشه‌بندی اولیه (Initial Clustering)**

در این مرحله، مشتریان بر اساس شباهت ویژگی‌های رفتاری گروه‌بندی می‌شوند.

برای شناخت چندبعدی رفتار مشتریان، از چارچوب LRRFM و سایر ویژگی‌های رفتاری استفاده خواهد شد.

هدف این مرحله کشف ساختار مشتریان و شناسایی گروه‌هایی با الگوهای خرید مشابه است.

**مرحله دوم: طبقه‌بندی (Classification)**

در این مرحله، هدف پیش‌بینی عدم خرید معتبر مشتریان در ۹۰ روز آینده است.

از آنجا که دیتاست برچسب آماده ریزش ندارد، برچسب هدف بر اساس رفتار آینده مشتری ساخته خواهد شد.

طبق تعریف عملیاتی پروژه:

- پنجره مشاهده: ۱۸۰ روز پیش از تاریخ مرجع
- افق پیش‌بینی: ۹۰ روز پس از تاریخ مرجع
- Churn = 1: عدم انجام خرید معتبر در ۹۰ روز آینده
- Churn = 0: انجام حداقل یک خرید معتبر در ۹۰ روز آینده

برای این کار، مدل‌های طبقه‌بندی آموزش داده می‌شوند و عملکرد آن‌ها با یک مدل خط مبنا و معیارهای ارزیابی مناسب مقایسه خواهد شد.

**مرحله سوم: خوشه‌بندی تکمیلی (Advanced Clustering)**

در این مرحله، مشتریان با استفاده از ویژگی‌های رفتاری و امتیازهای ریسک پیش‌بینی‌شده، مجدداً گروه‌بندی می‌شوند.

هدف، کشف گروه‌هایی است که از نظر رفتار خرید و سطح ریسک تفاوت دارند و می‌توانند برای طراحی راهبردهای متفاوت حفظ مشتری مورد استفاده قرار گیرند.

برای حفظ اعتبار تحلیلی، امتیازهای پیش‌بینی مورد استفاده باید خارج از نمونه تولید شده باشند.

### ۵. ارزیابی و تحلیل نتایج

عملکرد خوشه‌بندی با معیارهایی مانند Silhouette Score، Davies-Bouldin Index و بررسی پایداری و تفسیرپذیری خوشه‌ها ارزیابی خواهد شد.

عملکرد طبقه‌بندی نیز با معیارهایی مانند Precision، Recall، F1-score، PR-AUC، ROC-AUC و Confusion Matrix بررسی می‌شود.

همچنین ارزیابی زمانی و جلوگیری از نشت اطلاعات آینده از الزامات اساسی پروژه خواهند بود.

در ادامه، عواملی مانند شاخص‌های LRRFM، تغییرات رفتار خرید و سایر ویژگی‌های مهندسی‌شده از نظر ارتباط با عدم خرید آینده بررسی می‌شوند.

### ۶. کاربرد تجاری نتایج

یکی از خروجی‌های اصلی پروژه، شناسایی و رتبه‌بندی مشتریان بر اساس ریسک پیش‌بینی‌شده است.

با توجه به محدودیت ظرفیت تماس با ۵۰۰ مشتری، افراد دارای اولویت بالاتر برای ورود به فهرست کمپین انتخاب می‌شوند.

سپس با استفاده از اطلاعات رفتاری و گروه‌بندی مشتریان، راهبردهای متفاوتی مانند یادآوری خرید، پیشنهاد محصولات مرتبط یا پیشنهادهای تشویقی طراحی خواهد شد.

### ۷. تحلیل اقتصادی کمپین

برای بررسی امکان‌پذیری اقتصادی کمپین، هزینه‌های احتمالی اجرا، هزینه ارتباط با مشتری، هزینه تخفیف، سود افزایشی مورد انتظار و نرخ موفقیت لازم برای سربه‌سر شدن محاسبه خواهند شد.

همچنین تحلیل حساسیت انجام می‌شود تا مشخص شود تغییر هزینه کمپین یا ارزش اقتصادی مشتریان چه اثری بر تصمیم نهایی دارد.

از آنجا که اطلاعات مالی و اثربخشی واقعی کمپین مستقیماً در دیتاست وجود ندارند، محاسبات اقتصادی بر اساس فرضیات شفاف انجام خواهند شد.

### ۸. محدودیت‌ها و ریسک‌های پروژه

مهم‌ترین محدودیت‌های پروژه عبارت‌اند از:

- نبود برچسب آماده ریزش مشتری
- محدود بودن بازه زمانی دیتاست
- وجود شناسه‌های گمشده و تراکنش‌های غیرعادی
- نیاز به تعریف و اعتبارسنجی ویژگی‌های رفتاری
- احتمال نشت اطلاعات آینده
- نبود اطلاعات مستقیم سودآوری و تاریخچه کمپین‌ها
- محدودیت تماس با ۵۰۰ مشتری

همچنین عدم خرید در ۹۰ روز آینده یک تعریف عملیاتی برای این پروژه است و لزوماً به معنای از دست رفتن قطعی مشتری نیست.

پیش‌بینی ریسک نیز به‌تنهایی اثبات نمی‌کند که تماس یا تخفیف باعث جلوگیری از ریزش می‌شود. برای سنجش اثر واقعی مداخله بازاریابی، اجرای آزمایش کنترل‌شده مانند A/B Test ضروری خواهد بود.

### ۹. نتیجه مورد انتظار

نتیجه مورد انتظار پروژه، ایجاد یک چارچوب تحلیلی و تصمیم‌یار است که بتواند:

- ساختار رفتاری مشتریان را شناسایی کند.
- مشتریان در معرض عدم خرید را پیش‌بینی و اولویت‌بندی کند.
- عوامل مرتبط با رفتار خرید و عدم خرید را تحلیل کند.
- گروه‌های مناسب برای راهبردهای مختلف حفظ مشتری ایجاد کند.
- استفاده از ظرفیت محدود کمپین را هدفمندتر کند.
- امکان ارزیابی اقتصادی تصمیم‌های بازاریابی را فراهم آورد.

در نهایت، این پروژه تلاش می‌کند داده‌های خام تراکنش را به اطلاعات رفتاری، پیش‌بینی‌های قابل ارزیابی و پیشنهادهای عملی برای تصمیم‌گیری کسب‌وکار تبدیل کند.

</div>
