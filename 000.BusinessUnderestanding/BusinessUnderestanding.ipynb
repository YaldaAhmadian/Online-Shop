### سؤال ۱: مسئله اصلی کسب‌وکار چیست؟

**پاسخ:**
مسئله اصلی این است که فروشگاه نمی‌تواند به‌راحتی مشتریانی را که احتمال دارد خرید خود را متوقف کنند، پیش از وقوع ریزش شناسایی کند. در نتیجه، ممکن است فرصت حفظ مشتریان ارزشمند را از دست بدهد یا منابع بازاریابی را برای مشتریانی هزینه کند که نیازی به اقدام خاصی ندارند.

**چرا این سؤال را می‌پرسیم؟**

چون پیش از انتخاب الگوریتم یا تحلیل داده‌ها باید بدانیم دقیقاً چه مسئله‌ای را حل می‌کنیم. اگر مسئله کسب‌وکار مشخص نباشد، ممکن است مدلی بسازیم که از نظر فنی خوب باشد، اما برای فروشگاه کاربردی نداشته باشد.

### سؤال ۲: هدف اصلی کسب‌وکار از اجرای پروژه چیست؟

**پاسخ:**
هدف اصلی، شناسایی و اولویت‌بندی مشتریانی است که احتمال ریزش آن‌ها بالاست تا تیم فروش یا بازاریابی بتواند منابع خود را روی مشتریان مناسب متمرکز کند. هدف نهایی، کمک به حفظ مشتریان، کاهش ریزش و بهبود سودآوری فروشگاه است.

**چرا این سؤال را می‌پرسیم؟**

چون هدف پروژه تعیین می‌کند خروجی مدل چه کاربردی داشته باشد. در این پروژه، صرفاً پیش‌بینی ریزش کافی نیست؛ باید بتوان از پیش‌بینی برای تصمیم‌گیری درباره مشتریان استفاده کرد.

### سؤال ۳: چه افرادی از نتایج پروژه استفاده می‌کنند؟

**پاسخ:**
- **مدیر فروشگاه:** برای تصمیم‌گیری درباره راهبرد حفظ مشتریان و تخصیص بودجه.
- **تیم فروش و بازاریابی:** برای انتخاب مشتریانی که باید با آن‌ها ارتباط برقرار شود یا پیشنهاد ویژه دریافت کنند.
- **تیم داده‌کاوی:** برای آماده‌سازی داده‌ها، ساخت مدل، ارزیابی عملکرد و ارائه نتایج قابل‌تفسیر.

**چرا این سؤال را می‌پرسیم؟**

چون خروجی پروژه باید با نیاز استفاده‌کنندگان آن هماهنگ باشد. مدیر فروشگاه به شاخص‌های اقتصادی نیاز دارد، درحالی‌که تیم داده‌کاوی به معیارهای فنی و ارزیابی مدل نیازمند است.

### سؤال ۴: مشتریان هدف پروژه چه کسانی هستند؟

**پاسخ:**
مشتریان هدف، افرادی هستند که در داده‌های تراکنش فروشگاه سابقه خرید دارند و اطلاعات کافی برای محاسبه ویژگی‌های رفتاری آن‌ها موجود است.

در این پروژه، ابتدا باید مشتریان دارای شناسه معتبر را بررسی کنیم. مشتریانی که شناسه آن‌ها ثبت نشده است، ممکن است قابل‌اتصال به سابقه خرید مشخصی نباشند؛ بنابراین نمی‌توان بدون بررسی بیشتر، رفتار آن‌ها را به‌صورت فردی تحلیل کرد.

**چرا این سؤال را می‌پرسیم؟**

چون باید جامعه هدف مدل را مشخص کنیم. اگر معلوم نباشد مدل درباره کدام مشتریان پیش‌بینی می‌کند، نتایج آن قابل تفسیر و استفاده نخواهد بود.

### سؤال ۵: منظور ما از «ریزش مشتری» دقیقاً چیست؟

**پاسخ:**
در این پروژه، ریزش به‌صورت عملیاتی تعریف می‌شود: مشتری در بازه ۹۰ روزه پس از تاریخ مرجع، هیچ خرید معتبری ثبت نکند.

برای مثال، اگر تاریخ مرجع ۱ ژوئیه ۲۰۱۱ باشد، رفتار مشتری در ۹۰ روز بعد بررسی می‌شود. اگر در این بازه خرید معتبری نداشته باشد، برچسب ریزش برابر ۱ و در غیر این صورت برابر ۰ خواهد بود.

این تعریف، یک تعریف پژوهشی برای پروژه است و لزوماً به معنی اعلام رسمی قطع رابطه مشتری با فروشگاه نیست.

**چرا این سؤال را می‌پرسیم؟**

چون مدل یادگیری نظارت‌شده برای آموزش به یک متغیر هدف مشخص نیاز دارد. اگر ریزش را دقیق تعریف نکنیم، برچسب‌ها مبهم می‌شوند و ارزیابی مدل اعتبار کافی نخواهد داشت.

### سؤال ۶: چرا پیش‌بینی ریزش مشتری برای فروشگاه ارزشمند است؟

**پاسخ:**
پیش‌بینی ریزش می‌تواند به فروشگاه کمک کند مشتریان در معرض ریزش را زودتر شناسایی کند، فعالیت‌های بازاریابی را هدفمندتر انجام دهد و از صرف هزینه برای مشتریانی که احتمال ریزش پایینی دارند، در موارد غیرضروری جلوگیری کند.

بااین‌حال، صرف شناسایی مشتریان در معرض ریزش تضمین نمی‌کند که بتوان آن‌ها را حفظ کرد. اثربخشی اقدام‌های بازاریابی باید جداگانه ارزیابی شود.

**چرا این سؤال را می‌پرسیم؟**

چون باید مشخص شود خروجی پروژه چگونه به ارزش واقعی کسب‌وکار تبدیل می‌شود. یک مدل دقیق، به‌تنهایی معادل افزایش فروش یا سودآوری نیست.

### سؤال ۷: موفقیت پروژه از دید مدیر فروشگاه چگونه اندازه‌گیری می‌شود؟

**پاسخ:**
موفقیت کسب‌وکاری پروژه را می‌توان با شاخص‌های زیر ارزیابی کرد:

- تعداد و درصد مشتریان در معرض ریزش که به‌درستی شناسایی شده‌اند.
- میزان خرید مجدد مشتریانی که برای حفظ آن‌ها اقدام شده است.
- هزینه اجرای کمپین حفظ مشتری.
- سود حاصل از خریدهای مرتبط با کمپین، پس از کسر هزینه‌های مربوط.
- بازده سرمایه‌گذاری کمپین (ROI).

برای سنجش اثر واقعی کمپین، بهتر است مشتریان واجد شرایط به گروه مداخله و گروه کنترل تقسیم شوند و نتایج آن‌ها مقایسه شود.

**چرا این سؤال را می‌پرسیم؟**

چون معیار موفقیت از دید مدیر، صرفاً خوب بودن مدل نیست؛ بلکه این است که آیا استفاده از مدل به تصمیم بهتر و نتیجه اقتصادی مطلوب‌تر منجر می‌شود یا خیر.

### سؤال ۸: موفقیت پروژه از دید تیم فروش و بازاریابی چگونه اندازه‌گیری می‌شود؟

**پاسخ:**
از دید تیم فروش و بازاریابی، پروژه زمانی کاربردی است که فهرستی اولویت‌بندی‌شده از مشتریان در معرض ریزش ارائه کند، اطلاعات لازم برای پیگیری آن‌ها را در اختیار تیم قرار دهد و باعث شود زمان و بودجه بازاریابی هدفمندتر مصرف شود.

شاخص‌های مناسب عبارت‌اند از:

- تعداد مشتریان قابل‌پیگیری در فهرست پیشنهادی.
- نرخ پاسخ مشتریان به تماس یا پیشنهاد.
- نرخ خرید مجدد پس از اقدام بازاریابی.
- هزینه صرف‌شده به‌ازای هر مشتری حفظ‌شده.

این شاخص‌ها در حال حاضر معیارهای پیشنهادی هستند و برای محاسبه واقعی آن‌ها به داده‌های کمپین نیاز داریم.

**چرا این سؤال را می‌پرسیم؟**

چون ممکن است مدلی از نظر آماری مناسب باشد، اما خروجی آن برای تیم بازاریابی قابل‌استفاده نباشد. باید مشخص کنیم تیم دقیقاً چه اطلاعاتی برای اقدام عملی نیاز دارد.

### سؤال ۹: چه داده‌هایی در اختیار داریم و این داده‌ها از کجا آمده‌اند؟

**پاسخ:**
داده مورد استفاده پروژه، مجموعه‌داده عمومی Online Retail از مخزن UCI است. این مجموعه شامل تراکنش‌های یک فروشگاه اینترنتی بریتانیایی در بازه زمانی دسامبر ۲۰۱۰ تا دسامبر ۲۰۱۱ است.

ستون‌های اصلی داده عبارت‌اند از:

- `InvoiceNo`: شماره فاکتور
- `StockCode`: کد محصول
- `Description`: توضیحات محصول
- `Quantity`: تعداد اقلام
- `InvoiceDate`: تاریخ و زمان تراکنش
- `UnitPrice`: قیمت واحد
- `CustomerID`: شناسه مشتری
- `Country`: کشور مشتری

منبع داده:
https://archive.ics.uci.edu/dataset/352/online+retail

**چرا این سؤال را می‌پرسیم؟**

چون پیش از تحلیل باید منبع، ساختار، محدوده زمانی و محتوای داده را بشناسیم تا مشخص شود آیا داده موجود برای حل مسئله کافی است یا به اطلاعات دیگری نیاز داریم.

### سؤال ۱۰: داده‌های موجود چه محدودیت‌هایی دارند؟

**پاسخ:**
محدودیت‌های مهم عبارت‌اند از:

- وجود مقادیر گمشده در برخی ستون‌ها، به‌ویژه شناسه مشتری.
- وجود فاکتورهای لغوشده، مرجوعی‌ها یا مقادیر منفی در بعضی تراکنش‌ها.
- احتمال وجود رکوردهای تکراری.
- محدود بودن داده‌ها به یک فروشگاه و یک بازه تاریخی مشخص.
- نبود اطلاعات مستقیمی درباره هزینه‌های بازاریابی، سود واقعی مشتری و نتیجه کمپین‌های حفظ مشتری.
- نبود اطلاعات جمعیت‌شناختی کامل درباره مشتریان.
- محدود بودن بازه زمانی برای ارزیابی رفتار مشتری در دوره‌های طولانی‌تر.

وجود این موارد باید با بررسی واقعی داده تأیید و میزان آن‌ها در گزارش ثبت شود.

**چرا این سؤال را می‌پرسیم؟**

چون محدودیت‌های داده می‌توانند روی کیفیت برچسب‌ها، ویژگی‌ها، عملکرد مدل و امکان تعمیم نتایج اثر بگذارند. شناخت آن‌ها از نتیجه‌گیری‌های نادرست جلوگیری می‌کند.

### سؤال ۱۱: چه منابع و امکاناتی برای اجرای پروژه نیاز داریم؟

**پاسخ:**
منابع اصلی پروژه عبارت‌اند از:

- مجموعه‌داده عمومی Online Retail.
- محیط برنامه‌نویسی Python و ابزارهایی مانند Jupyter Notebook.
- کتابخانه‌های pandas و NumPy برای پردازش داده‌ها.
- کتابخانه‌های matplotlib و seaborn برای مصورسازی.
- کتابخانه scikit-learn برای ساخت و ارزیابی مدل‌ها.
- مستندات علمی مرتبط با پیش‌بینی ریزش مشتری.
- زمان کافی برای پاک‌سازی داده، ساخت ویژگی‌ها، مدل‌سازی و ارزیابی.

**چرا این سؤال را می‌پرسیم؟**

چون باید مطمئن شویم پروژه با امکانات موجود قابل‌اجراست و ابزارها و مهارت‌های موردنیاز از ابتدا مشخص شده‌اند.

### سؤال ۱۲: چه اطلاعاتی برای حل مسئله نداریم و ممکن است به آن‌ها نیاز داشته باشیم؟

**پاسخ:**
داده موجود رفتار خرید را نشان می‌دهد، اما اطلاعات زیر را در اختیار ما قرار نمی‌دهد یا دست‌کم به‌صورت مستقیم در ستون‌های اصلی آن وجود ندارد:

- هزینه تماس با مشتری یا اجرای کمپین.
- میزان تخفیف ارائه‌شده به هر مشتری.
- سود خالص حاصل از خرید هر مشتری.
- سابقه تماس‌های بازاریابی و پاسخ مشتریان.
- نتیجه کمپین‌های قبلی.
- علت واقعی توقف خرید مشتری.

بنابراین، می‌توانیم احتمال ریزش را بر اساس رفتار خرید تخمین بزنیم؛ اما بدون داده تکمیلی نمی‌توانیم علت ریزش یا سود واقعی کمپین را به‌طور قطعی تعیین کنیم.

**چرا این سؤال را می‌پرسیم؟**

چون باید تفاوت بین آنچه واقعاً می‌دانیم و آنچه صرفاً فرض می‌کنیم مشخص باشد. این کار از ادعاهای غیرقابل‌دفاع درباره نتایج پروژه جلوگیری می‌کند.

### سؤال ۱۳: چه ریسک‌هایی ممکن است پروژه را با مشکل مواجه کنند؟

**پاسخ:**
ریسک‌های اصلی عبارت‌اند از:

- تعریف نامناسب یا مبهم ریزش.
- نشت اطلاعات آینده به ویژگی‌های آموزشی مدل (Data Leakage).
- نامتوازن بودن تعداد مشتریان ریزشی و غیرریزشـی.
- کیفیت پایین یا ناقص بودن داده‌ها.
- ارزیابی مدل با روش نامناسب.
- تعمیم نتایج تاریخی یک فروشگاه به فروشگاه‌های دیگر.
- انتخاب آستانه تصمیم‌گیری بدون توجه به هزینه خطاها.
- فرض گرفتن اینکه شناسایی مشتری در معرض ریزش، لزوماً باعث حفظ او می‌شود.

**چرا این سؤال را می‌پرسیم؟**

چون شناسایی ریسک‌ها در ابتدای پروژه کمک می‌کند برای کنترل آن‌ها برنامه داشته باشیم و از نتایج گمراه‌کننده جلوگیری کنیم.

### سؤال ۱۴: چه فرض‌هایی در این پروژه داریم؟

**پاسخ:**
فرض‌های اصلی عبارت‌اند از:

- سابقه تراکنش‌های ثبت‌شده، نماینده‌ای قابل‌قبول از رفتار خرید مشتری است.
- مشتریان دارای شناسه معتبر را می‌توان در تراکنش‌های مختلف به یکدیگر مرتبط کرد.
- نبود خرید معتبر در ۹۰ روز آینده، تعریف عملیاتی مناسبی برای ریزش در این پروژه است.
- ویژگی‌های محاسبه‌شده از گذشته می‌توانند تا حدی درباره رفتار خرید آینده اطلاعات بدهند.
- رفتار تاریخی مشتریان برای بررسی الگوهای ریزش مفید است، اما لزوماً رفتار آینده یا همه فروشگاه‌ها را به‌طور کامل پیش‌بینی نمی‌کند.

**چرا این سؤال را می‌پرسیم؟**

چون هر پروژه داده‌کاوی بر مجموعه‌ای از فرض‌ها استوار است. ثبت فرض‌ها کمک می‌کند بدانیم نتایج تحت چه شرایطی معتبرند و کجا باید با احتیاط تفسیر شوند.

### سؤال ۱۵: هدف اصلی داده‌کاوی چیست و مسئله از چه نوعی است؟

**پاسخ:**
هدف اصلی، ساخت مدلی است که با استفاده از سابقه خرید مشتری، احتمال ریزش او را تخمین بزند یا مشخص کند که آیا در بازه تعریف‌شده در معرض ریزش قرار می‌گیرد یا خیر.

این مسئله از نوع یادگیری نظارت‌شده (Supervised Learning) و در صورت تعریف خروجی صفر و یک، یک مسئله طبقه‌بندی دودویی (Binary Classification) است.

- برچسب ۱: مشتری در ۹۰ روز پس از تاریخ مرجع خرید معتبر ندارد.
- برچسب ۰: مشتری در این بازه دست‌کم یک خرید معتبر دارد.

**چرا این سؤال را می‌پرسیم؟**

چون نوع مسئله تعیین می‌کند از چه الگوریتم‌ها، چه ساختار داده‌ای و چه معیارهایی برای ارزیابی استفاده کنیم.

### سؤال ۱۶: مدل باید از چه اطلاعاتی برای پیش‌بینی استفاده کند؟

**پاسخ:**
ویژگی‌های مدل باید فقط از اطلاعاتی ساخته شوند که تا تاریخ مرجع در دسترس بوده‌اند. ویژگی‌های پیشنهادی عبارت‌اند از:

- `Recency`: تعداد روزهای گذشته از آخرین خرید معتبر.
- `Frequency`: تعداد خریدها یا فاکتورهای معتبر در بازه گذشته.
- `Monetary`: مجموع ارزش خرید در بازه گذشته.
- میانگین ارزش هر فاکتور.
- میانگین فاصله زمانی بین خریدها.
- تعداد روزهای فعال خرید.
- تعداد محصولات منحصربه‌فرد خریداری‌شده.
- تغییر تعداد یا ارزش خرید نسبت به بازه قبلی.
- مدت‌زمان سپری‌شده از اولین خرید ثبت‌شده تا تاریخ مرجع.

تعریف دقیق هر ویژگی، نحوه برخورد با مرجوعی‌ها و بازه زمانی محاسبه باید در مرحله آماده‌سازی داده ثبت شود.

**چرا این سؤال را می‌پرسیم؟**

چون ویژگی‌ها اطلاعاتی هستند که مدل بر اساس آن‌ها تصمیم می‌گیرد. انتخاب ویژگی‌های مرتبط می‌تواند قدرت پیش‌بینی را افزایش دهد، درحالی‌که ویژگی‌های نامناسب ممکن است باعث خطا یا نشت اطلاعات شوند.

### سؤال ۱۷: مدل را چگونه آموزش می‌دهیم و ارزیابی می‌کنیم؟

**پاسخ:**
برای کاهش خطر نشت اطلاعات زمانی، از چند تاریخ مرجع استفاده می‌کنیم. برای نمونه:

- تاریخ مرجع اول: ۱ ژوئن ۲۰۱۱.
- تاریخ مرجع دوم: ۱ ژوئیه ۲۰۱۱.
- تاریخ مرجع سوم: ۱ اوت ۲۰۱۱.
- تاریخ مرجع چهارم: ۱ سپتامبر ۲۰۱۱.

برای هر تاریخ مرجع، ویژگی‌ها از ۱۸۰ روز پیش از آن محاسبه می‌شوند و برچسب با بررسی ۹۰ روز پس از آن ساخته می‌شود.

مدل باید روی دوره‌های قدیمی‌تر آموزش ببیند و روی دوره‌ای جدیدتر که در آموزش استفاده نشده است ارزیابی شود. همچنین باید اطمینان حاصل کنیم که اطلاعات مربوط به دوره هدف وارد ویژگی‌های پیش‌بینی نشده باشد.

**چرا این سؤال را می‌پرسیم؟**

چون مدل باید شرایط واقعی پیش‌بینی آینده را شبیه‌سازی کند. اگر اطلاعات آینده وارد آموزش شود، عملکرد مدل به‌صورت غیرواقعی خوب به نظر می‌رسد.

### سؤال ۱۸: از چه الگوریتم‌هایی برای مقایسه استفاده می‌کنیم؟

**پاسخ:**
مدل‌های پیشنهادی عبارت‌اند از:

- مدل پایه مبتنی بر RFM: استفاده از تازگی، تعداد و ارزش خرید برای ایجاد یک معیار مقایسه ساده.
- Logistic Regression: یک مدل پایه قابل‌تفسیر برای طبقه‌بندی.
- Decision Tree: برای بررسی قواعد تصمیم‌گیری.
- Random Forest: برای یادگیری روابط غیرخطی و ترکیب چند درخت تصمیم.
- Gradient Boosting: برای بررسی عملکرد مدل‌های تقویتی.

مدل‌ها باید با تقسیم‌بندی زمانی یکسان و مجموعه ویژگی‌های قابل‌مقایسه ارزیابی شوند. انتخاب مدل نهایی نباید صرفاً بر اساس پیچیدگی یا شهرت الگوریتم باشد.

**چرا این سؤال را می‌پرسیم؟**

چون بدون مقایسه با مدل پایه نمی‌دانیم روش پیچیده‌تر واقعاً ارزش افزوده ایجاد کرده است یا خیر.

### سؤال ۱۹: عملکرد مدل را با چه معیارهایی می‌سنجیم؟

**پاسخ:**
معیارهای اصلی ارزیابی عبارت‌اند از:

- Precision: از مشتریانی که مدل ریزشی پیش‌بینی کرده، چه نسبتی واقعاً ریزش داشته‌اند؟
- Recall: از تمام مشتریانی که واقعاً ریزش داشته‌اند، چه نسبتی شناسایی شده‌اند؟
- F1-score: میانگین هماهنگ Precision و Recall.
- PR-AUC: ارزیابی کیفیت رتبه‌بندی در شرایطی که ممکن است کلاس‌ها نامتوازن باشند.
- ROC-AUC: ارزیابی توانایی مدل در تفکیک دو گروه در آستانه‌های مختلف.
- Confusion Matrix: نمایش تعداد پیش‌بینی‌های درست و خطاها.

همچنین باید عملکرد مدل را با مدل پایه مقایسه کنیم. Accuracy به‌تنهایی کافی نیست؛ به‌ویژه اگر تعداد مشتریان ریزشی و غیرریزشـی متفاوت باشد.

**چرا این سؤال را می‌پرسیم؟**

چون هیچ معیار واحدی تمام جنبه‌های عملکرد مدل را نشان نمی‌دهد. انتخاب معیار مناسب به هدف تجاری و هزینه خطاها وابسته است.

### سؤال ۲۰: خطاهای مدل چه هستند و کدام‌یک اهمیت بیشتری دارد؟

**پاسخ:**
دو خطای اصلی عبارت‌اند از:

- False Positive (FP): مدل مشتری را ریزشی پیش‌بینی می‌کند، اما مشتری در بازه ۹۰ روزه خرید معتبر دارد. در صورت تماس یا ارائه تخفیف، ممکن است بخشی از منابع بازاریابی بدون نیاز مصرف شود.
- False Negative (FN): مدل مشتری را غیرریزشـی پیش‌بینی می‌کند، اما مشتری در بازه موردنظر خرید معتبری ندارد. در این حالت، ممکن است فرصتی برای اقدام پیشگیرانه از دست برود.

اینکه کدام خطا پرهزینه‌تر است، به هزینه کمپین، سود احتمالی مشتری و امکان حفظ او بستگی دارد؛ بنابراین نباید بدون اطلاعات اقتصادی، یکی را قطعاً مهم‌تر بدانیم.

**چرا این سؤال را می‌پرسیم؟**

چون آستانه تصمیم‌گیری مدل باید با پیامدهای واقعی خطاها هماهنگ باشد. مدلی با Recall بالاتر ممکن است مشتریان بیشتری را شناسایی کند، اما تعداد بیشتری هشدار اشتباه نیز ایجاد کند.

### سؤال ۲۱: خروجی نهایی مدل باید چه چیزی باشد؟

**پاسخ:**
خروجی مدل باید شامل یک فهرست اولویت‌بندی‌شده از مشتریان باشد که برای هر مشتری، اطلاعات زیر را ارائه کند:

- شناسه مشتری.
- احتمال پیش‌بینی‌شده ریزش، در صورت استفاده از مدل دارای خروجی احتمالاتی.
- برچسب پیش‌بینی‌شده در آستانه انتخاب‌شده.
- چند ویژگی رفتاری مهم، مانند تازگی خرید و تعداد خریدها.
- اولویت پیشنهادی برای بررسی توسط تیم بازاریابی.

احتمال پیش‌بینی‌شده باید در صورت امکان از نظر کالیبراسیون بررسی شود؛ این احتمال لزوماً احتمال واقعی و قطعی ریزش نیست.

**چرا این سؤال را می‌پرسیم؟**

چون خروجی قابل‌استفاده برای کسب‌وکار باید فراتر از یک عدد ارزیابی باشد و به تصمیم‌گیری درباره مشتریان کمک کند.

### سؤال ۲۲: آیا خوشه‌بندی مشتریان نیز در پروژه کاربرد دارد؟

**پاسخ:**
بله، خوشه‌بندی می‌تواند به‌عنوان تحلیل تکمیلی انجام شود. برای مثال، می‌توان مشتریان را بر اساس الگوهای خرید به گروه‌هایی مانند مشتریان پرتکرار، کم‌خرید یا دارای ارزش خرید بالا تقسیم کرد.

سپس می‌توان نرخ ریزش را بین خوشه‌ها مقایسه کرد. بااین‌حال، خوشه‌بندی جایگزین مدل پیش‌بینی ریزش نیست و وجود یک خوشه به‌تنهایی به معنی ریزشی بودن اعضای آن نیست.

**چرا این سؤال را می‌پرسیم؟**

چون شناخت گروه‌های رفتاری ممکن است به طراحی راهبردهای متفاوت برای حفظ مشتریان کمک کند، اما باید از هدف اصلی پروژه جدا نگه داشته شود.

### سؤال ۲۳: از چه زمانی باید پیش‌بینی را انجام دهیم و اطلاعات تا چه زمانی مجاز است؟

**پاسخ:**
پیش‌بینی باید در یک تاریخ مرجع مشخص انجام شود. فقط اطلاعات موجود تا همان تاریخ می‌توانند برای ساخت ویژگی‌ها استفاده شوند.

برای مثال، اگر تاریخ مرجع ۱ ژوئیه ۲۰۱۱ باشد، ویژگی‌ها از سابقه ۱۸۰ روز پیش از آن ساخته می‌شوند و رفتار ۹۰ روز بعد فقط برای تعیین برچسب استفاده خواهد شد.

**چرا این سؤال را می‌پرسیم؟**

چون در مسئله پیش‌بینی ریزش، استفاده از اطلاعاتی که در زمان واقعی پیش‌بینی هنوز در دسترس نیستند، باعث نشت اطلاعات و ارزیابی غیرواقعی می‌شود.

### سؤال ۲۴: چه معیارهایی برای پذیرش پروژه در نظر می‌گیریم؟

**پاسخ:**
پروژه زمانی قابل‌قبول است که:

- تعریف ریزش و روش ساخت برچسب مستند باشد.
- کیفیت داده‌ها و مقادیر گمشده بررسی و گزارش شده باشند.
- ویژگی‌ها بدون نشت اطلاعات زمانی ساخته شده باشند.
- مدل‌ها با روش ارزیابی زمانی و معیارهای یکسان مقایسه شده باشند.
- عملکرد مدل از یک معیار پایه مشخص بهتر باشد یا دلایل نبود بهبود مستند شود.
- نتایج قابل‌بازتولید باشند.
- محدودیت‌ها و ریسک‌های استفاده از نتایج توضیح داده شوند.
- خروجی نهایی برای استفاده در تصمیم‌گیری کسب‌وکار قابل‌فهم باشد.

آستانه عددی قابل‌قبول برای معیارهای مدل باید پس از بررسی داده و تعیین نیازهای پروژه مشخص شود؛ در حال حاضر مبنای کافی برای اعلام یک عدد قطعی نداریم.

**چرا این سؤال را می‌پرسیم؟**

چون باید پیش از پایان پروژه بدانیم چه شواهدی نشان می‌دهند کار انجام‌شده قابل‌قبول است. معیار پذیرش نباید صرفاً به یک عدد مانند Accuracy محدود شود.

### سؤال ۲۵: چه مراحل و خروجی‌هایی برای اجرای پروژه لازم است؟

**پاسخ:**
برنامه پروژه بر اساس مراحل CRISP-DM به شکل زیر است:

1. درک کسب‌وکار (Business Understanding): تعریف مسئله، هدف، جامعه هدف و معیارهای موفقیت.
2. درک داده‌ها (Data Understanding): بررسی ساختار داده، کیفیت، مقادیر گمشده، تراکنش‌های غیرعادی و توزیع مشتریان.
3. آماده‌سازی داده‌ها (Data Preparation): پاک‌سازی، تعریف تراکنش معتبر، ساخت ویژگی‌های رفتاری و ایجاد برچسب ریزش.
4. مدل‌سازی (Modeling): آموزش و مقایسه مدل‌های پایه و طبقه‌بندی.
5. ارزیابی (Evaluation): بررسی معیارهای فنی، نشت اطلاعات، خطاها و تناسب نتایج با هدف کسب‌وکار.
6. ارائه یا استقرار (Deployment): تهیه گزارش مدیریتی و فهرست اولویت‌بندی‌شده مشتریان؛ در این پروژه، استقرار عملیاتی واقعی جزو خروجی قطعی محسوب نمی‌شود.

**چرا این سؤال را می‌پرسیم؟**

چون تقسیم پروژه به مراحل مشخص، اجرای کار را منظم‌تر می‌کند و باعث می‌شود خروجی هر مرحله مبنای مرحله بعد باشد.

### سؤال ۲۶: چه محدودیت‌هایی برای زمان، منابع و دامنه پروژه داریم؟

**پاسخ:**
این پروژه بر یک مجموعه‌داده عمومی و تاریخی متکی است؛ بنابراین، به اطلاعات داخلی فروشگاه، داده‌های واقعی کمپین و هزینه‌های دقیق کسب‌وکار دسترسی مستقیم نداریم.

دامنه اصلی پروژه باید روی پیش‌بینی ریزش از روی رفتار خرید متمرکز بماند. تحلیل سودآوری کمپین، خوشه‌بندی مشتریان و پیشنهاد راهبردهای بازاریابی می‌توانند بخش‌های تکمیلی باشند، اما نباید بدون داده کافی به‌عنوان نتیجه قطعی ارائه شوند.

**چرا این سؤال را می‌پرسیم؟**

چون مشخص کردن دامنه و محدودیت‌ها مانع گسترش بی‌رویه پروژه می‌شود و کمک می‌کند زمان و منابع روی هدف اصلی متمرکز بمانند.

### سؤال ۲۷: چگونه از صحت و قابلیت بازتولید نتایج مطمئن می‌شویم؟

**پاسخ:**
برای بازتولیدپذیری و کنترل کیفیت باید:

- کدهای پاک‌سازی و ساخت ویژگی‌ها مستند باشند.
- تاریخ‌های مرجع و تعریف برچسب‌ها ثبت شوند.
- داده آموزش و ارزیابی به‌صورت زمانی تفکیک شود.
- تمام مراحل پیش‌پردازش فقط با اطلاعات مجاز دوره آموزش برازش شوند.
- نسخه کتابخانه‌ها و تنظیمات مهم مدل ثبت شوند.
- نتایج مدل‌ها و معیارهای ارزیابی ذخیره شوند.
- محدودیت‌ها و تصمیم‌های مربوط به داده‌های گمشده، مرجوعی‌ها و تراکنش‌های لغوشده مستند باشند.

**چرا این سؤال را می‌پرسیم؟**

چون نتیجه علمی زمانی قابل‌اعتمادتر است که بتوان مراحل تولید آن را دوباره اجرا کرد و فهمید هر تصمیم چگونه بر نتیجه اثر گذاشته است.
