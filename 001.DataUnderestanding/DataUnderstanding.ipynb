```python
invoice_line_counts = df.groupby("InvoiceNo").size()

print("Invoice Line Statistics:")
print(invoice_line_counts.describe())
```

```python
invoice_customer_counts = (
    df.groupby("InvoiceNo")["CustomerID"]
    .nunique()
)

print("Invoices linked to one customer:")
print(invoice_customer_counts.eq(1).sum())

print("\nInvoices linked to multiple customers:")
print(invoice_customer_counts.gt(1).sum())

print("\nInvoices without a known customer:")
print(invoice_customer_counts.eq(0).sum())
```

# سؤال پنجم: بررسی واحد هر رکورد و روابط بین موجودیت‌ها (Record Granularity & Entity Relationships)

## هدف

هدف از این بخش، تعیین سطح جزئیات هر رکورد و شناسایی ارتباط بین موجودیت‌های اصلی مجموعه‌داده، شامل مشتری، فاکتور و محصول است.

شناخت واحد هر رکورد اهمیت زیادی دارد؛ زیرا مشخص می‌کند هر ردیف چه چیزی را توصیف می‌کند و برای انجام تحلیل‌های بعدی باید اطلاعات را در چه سطحی بررسی کنیم.

## واحد هر رکورد (Record Granularity)

با توجه به ساختار مجموعه‌داده Online Retail، هر ردیف نشان‌دهنده یک قلم کالا در یک فاکتور است.

به عبارت دیگر، هر رکورد اطلاعات مربوط به یک محصول ثبت‌شده در یک فاکتور را شامل می‌شود؛ بنابراین هر ردیف الزاماً معادل یک فاکتور کامل یا یک مشتری مستقل نیست.

برای مثال، اگر یک مشتری در یک فاکتور سه محصول متفاوت خریداری کرده باشد، اطلاعات آن فاکتور می‌تواند در سه ردیف جداگانه ثبت شود.

## موجودیت‌های اصلی مجموعه‌داده

سه موجودیت اصلی قابل شناسایی در این مجموعه‌داده عبارت‌اند از:

### ۱. مشتری (Customer)

هر مشتری با استفاده از ستون `CustomerID` شناسایی می‌شود.

یک مشتری می‌تواند در طول زمان چندین فاکتور داشته باشد. بنابراین، شناسه یک مشتری ممکن است در تعداد زیادی از ردیف‌های مجموعه‌داده تکرار شود.

### ۲. فاکتور (Invoice)

هر فاکتور با استفاده از ستون `InvoiceNo` شناسایی می‌شود.

یک فاکتور می‌تواند شامل چندین ردیف کالا باشد؛ بنابراین مقدار `InvoiceNo` ممکن است در چندین رکورد تکرار شود.

### ۳. محصول (Product)

هر محصول با استفاده از ستون `StockCode` شناسایی می‌شود.

یک محصول می‌تواند در فاکتورهای مختلف و توسط مشتریان متفاوت خریداری شود؛ بنابراین مقدار `StockCode` نیز می‌تواند در رکوردهای متعددی تکرار شود.

## ارتباط بین موجودیت‌ها

روابط اصلی بین موجودیت‌ها به شکل زیر قابل توصیف هستند:

- هر مشتری می‌تواند چندین فاکتور داشته باشد.
- هر فاکتور می‌تواند شامل چندین ردیف محصول باشد.
- هر محصول می‌تواند در فاکتورهای مختلف ثبت شود.
- هر فاکتور در حالت معمول به یک مشتری مشخص مرتبط است؛ البته وجود مقادیر گمشده در `CustomerID` باعث می‌شود ارتباط برخی فاکتورها با مشتری قابل شناسایی نباشد.

در نتیجه، ارتباط مشتری و فاکتور از نوع یک‌به‌چند است و ارتباط فاکتور و محصول در سطح خرید، از طریق ردیف‌های فاکتور ثبت می‌شود.

## اهمیت سطح جزئیات داده در پروژه

از آنجا که هدف پروژه تحلیل رفتار مشتریان و بررسی ریزش آن‌هاست، باید توجه داشته باشیم که داده اولیه در سطح ردیف کالا ثبت شده است، نه در سطح مشتری.

بنابراین، برای تحلیل رفتار هر مشتری باید اطلاعات مربوط به ردیف‌های متعدد او در طول زمان به شکل مناسبی بررسی و در مراحل بعدی، در سطح مشتری خلاصه‌سازی شوند.

برای مثال:

- تعداد خریدهای مشتری باید بر اساس فاکتورهای متمایز محاسبه شود، نه تعداد ردیف‌های کالا.
- میزان خرید باید با توجه به تعداد و قیمت محصولات بررسی شود.
- تنوع محصولات خریداری‌شده باید بر اساس شناسه محصولات متمایز بررسی شود.
- فاصله زمانی خریدها باید با استفاده از تاریخ فاکتورهای مشتری محاسبه شود.

از آنجا که یک مشتری می‌تواند چندین فاکتور داشته باشد و هر فاکتور نیز می‌تواند شامل چندین ردیف کالا باشد، تعداد رکوردهای دیتاست با تعداد فاکتورها یا تعداد مشتریان برابر نیست.

رکورد تکراری (Duplicate Record) به رکوردی گفته می‌شود که مقادیر آن در تمام ستون‌های مورد بررسی با رکورد دیگری یکسان باشد.

برای بررسی این مسئله باید دو مفهوم را از هم جدا کنیم:

**۱. تکرار کامل رکورد (Full Duplicate)**

در این حالت، تمام مقادیر دو یا چند ردیف در ستون‌های دیتاست یکسان هستند؛ یعنی InvoiceNo، StockCode، Description، Quantity، InvoiceDate، UnitPrice، CustomerID و Country همگی برابرند.

**۲. تکرار در سطح بخشی از اطلاعات**

ممکن است چند ردیف دارای InvoiceNo یا StockCode یکسان باشند، اما سایر ستون‌های آن‌ها متفاوت باشد. چنین حالتی لزوماً Duplicate محسوب نمی‌شود؛ زیرا ممکن است نشان‌دهنده چند قلم کالا در یک فاکتور یا فروش یک محصول در زمان‌های مختلف باشد.

### اهمیت رکوردهای تکراری در پروژه Churn

وجود رکوردهای کاملاً تکراری می‌تواند باعث شود تعداد خریدها، تعداد اقلام فروخته‌شده و مبلغ فروش بیشتر از مقدار واقعی محاسبه شود.

از آنجا که در پروژه ما ویژگی‌هایی مانند تعداد خرید، میزان خرید و مجموع هزینه‌کرد مشتری اهمیت دارند، وجود رکوردهای تکراری می‌تواند بر محاسبه این ویژگی‌ها و در نهایت بر تحلیل رفتار مشتری تأثیر بگذارد.

### جمع‌بندی

در این مرحله باید تعداد رکوردهای کاملاً تکراری و درصد آن‌ها نسبت به کل داده‌ها مشخص شود. همچنین باید نمونه‌هایی از این رکوردها بررسی شوند تا مشخص شود تکرارها در چه سطحی اتفاق افتاده‌اند.

در فاز Data Understanding هیچ رکوردی حذف نمی‌شود؛ تصمیم‌گیری درباره حذف یا نگهداری رکوردهای تکراری در فاز Data Preparation انجام خواهد شد.

### سؤال ۹: داده‌ها چه بازه زمانی‌ای را پوشش می‌دهند و آیا این بازه برای تعریف پنجره‌های مشاهده و برچسب‌گذاری کافی است؟

**چرایی و دلیل مطرح‌شدن سؤال:**

در مسئله ریزش مشتری، زمان نقش اساسی دارد. برای تحلیل رفتار گذشته مشتری و بررسی خریدهای آینده، باید داده‌ها پوشش زمانی کافی داشته باشند. همچنین محدودیت ابتدا و انتهای بازه داده می‌تواند تعداد تاریخ‌های مرجع قابل استفاده را تعیین کند.

**هدف:**

شناخت محدوده زمانی داده‌ها، بررسی پیوستگی زمانی و ارزیابی امکان تعریف پنجره مشاهده ۱۸۰ روزه و پنجره برچسب‌گذاری ۹۰ روزه.

**فایده برای پروژه:**

مشخص می‌شود کدام تاریخ‌های مرجع از نظر زمانی قابل بررسی هستند و آیا داده‌ها امکان ارزیابی معتبر و بدون استفاده از اطلاعات آینده را فراهم می‌کنند.

</div>

```python
invoice_dates = pd.to_datetime(
    df["InvoiceDate"],
    format="%m/%d/%y %H:%M"
)

first_transaction = invoice_dates.min()
last_transaction = invoice_dates.max()

print("First transaction:", first_transaction)
print("Last transaction:", last_transaction)
```

```python
observation_days = 180
churn_days = 90

coverage_days = (last_transaction - first_transaction).total_seconds() / 86_400
required_days = observation_days + churn_days
is_sufficient = coverage_days >= required_days

print(f"Required days: {required_days}")
print(f"Available days: {coverage_days:.2f}")
print(f"Is sufficient: {is_sufficient}")
```

```python
earliest_reference_date = (
    first_transaction + pd.Timedelta(days=observation_days)
)

latest_reference_date = (
    last_transaction - pd.Timedelta(days=churn_days)
)

print("Earliest Reference Date:", earliest_reference_date)
print("Latest Reference Date:", latest_reference_date)
```

```python
monthly_transactions = (
    invoice_dates
    .dt.to_period("M")
    .value_counts()
    .sort_index()
)

print(monthly_transactions)
```

### بررسی پوشش زمانی مجموعه‌داده

یکی از پیش‌نیازهای تحلیل رفتار مشتری و تعریف ریزش، وجود اطلاعات زمانی کافی از تراکنش‌های مشتریان است. در پروژه حاضر، برای تعریف عملیاتی Churn به دو بازه زمانی نیاز داریم:

- **Observation Window:** یک بازه ۱۸۰ روزه برای بررسی سابقه و رفتار خرید مشتری.
- **Churn Window:** یک بازه ۹۰ روزه پس از پایان دوره مشاهده برای بررسی ادامه یا عدم ادامه فعالیت خرید مشتری.

بنابراین، برای ساخت یک نمونه قابل ارزیابی، حداقل به ۲۷۰ روز اطلاعات زمانی پیوسته نیاز داریم.

بررسی اولیه مجموعه‌داده Online Retail نشان می‌دهد که اطلاعات تراکنش‌ها از ۱ دسامبر ۲۰۱۰ تا ۹ دسامبر ۲۰۱۱ ثبت شده‌اند. در نتیجه، دیتاست تقریباً ۳۷۳ روز پوشش زمانی دارد.

با توجه به اینکه طول دوره موجود بیشتر از مجموع ۲۷۰ روز موردنیاز است، از نظر پوشش کلی زمانی امکان تعریف این دو پنجره وجود دارد.

با این حال، صرف کافی بودن طول کلی دیتاست به این معنا نیست که هر تاریخی می‌تواند به‌عنوان Reference Date انتخاب شود. تاریخ مرجع باید به‌گونه‌ای انتخاب شود که هم ۱۸۰ روز اطلاعات پیش از آن و هم ۹۰ روز اطلاعات پس از آن در دیتاست موجود باشد.

بنابراین، انتخاب تاریخ مرجع در این پروژه اهمیت زیادی دارد و باید محدودیت‌های ابتدا و انتهای بازه زمانی دیتاست در نظر گرفته شوند.

در مجموع، پوشش زمانی مجموعه‌داده از نظر طول کلی برای تعریف پنجره‌های ۱۸۰ و ۹۰ روزه کافی است؛ اما امکان استفاده از تمام تاریخ‌های موجود به‌عنوان تاریخ مرجع وجود ندارد.

### سؤال ۱۵: فعالیت خرید مشتریان در طول زمان چه روندها و الگوهای زمانی‌ای دارد؟

**چرایی و دلیل مطرح‌شدن سؤال:**

رفتار خرید مشتریان در طول زمان ثابت نیست. ممکن است میزان خرید در روزها، ماه‌ها یا دوره‌های خاص تغییر کند. این تغییرات می‌توانند ناشی از فصل، مناسبت‌ها، چرخه طبیعی خرید یا تغییر رفتار مشتری باشند.

**هدف:**

شناخت روند زمانی تراکنش‌ها و بررسی تغییرات حجم خرید و فعالیت مشتریان در بازه‌های زمانی مختلف.

**فایده برای پروژه:**

امکان تشخیص الگوهای زمانی و درک بهتر رفتار خرید فراهم می‌شود. همچنین از اشتباه‌گرفتن تغییرات فصلی و طبیعی فروش با نشانه‌های ریزش مشتری جلوگیری خواهد شد.

```python
invoice_dates = pd.to_datetime(
    df["InvoiceDate"],
    format="%m/%d/%y %H:%M"
)

print("First transaction:", invoice_dates.min())
print("Last transaction:", invoice_dates.max())
print("Date range:", invoice_dates.max() - invoice_dates.min())
```

```python
monthly_records = (
    df.assign(InvoiceDateParsed=invoice_dates)
    .groupby(
        pd.Grouper(
            key="InvoiceDateParsed",
            freq="MS"
        )
    )
    .size()
)

print(monthly_records)
```

```python
monthly_invoices = (
    df.assign(InvoiceDateParsed=invoice_dates)
    .groupby(
        pd.Grouper(
            key="InvoiceDateParsed",
            freq="MS"
        )
    )["InvoiceNo"]
    .nunique()
)

print(monthly_invoices)
```

```python
monthly_summary = (
    df.assign(InvoiceDateParsed=invoice_dates)
    .groupby(
        pd.Grouper(
            key="InvoiceDateParsed",
            freq="MS"
        )
    )
    .agg(
        RecordCount=("InvoiceNo", "size"),
        InvoiceCount=("InvoiceNo", "nunique")
    )
)

monthly_summary
```

```python
weekday_names = {
    0: "Monday",
    1: "Tuesday",
    2: "Wednesday",
    3: "Thursday",
    4: "Friday",
    5: "Saturday",
    6: "Sunday"
}

weekday_records = (
    invoice_dates.dt.dayofweek
    .map(weekday_names)
    .value_counts()
)

weekday_records
```

```python
import matplotlib.pyplot as plt

monthly_records.plot(
    kind="line",
    marker="o",
    figsize=(12, 5)
)

plt.title("Monthly Transaction Records")
plt.xlabel("Month")
plt.ylabel("Number of Records")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

### بررسی وجود برچسب Churn و شواهد رفتاری برای تعریف آن

یکی از اهداف اصلی پروژه Online Retail، شناسایی مشتریانی است که احتمال دارد دیگر از فروشگاه خرید نکنند.

#### ۱. بررسی وجود برچسب Churn در دیتاست

دیتاست اصلی Online Retail شامل هشت ستون زیر است:

- InvoiceNo
- StockCode
- Description
- Quantity
- InvoiceDate
- UnitPrice
- CustomerID
- Country

در میان این ستون‌ها، هیچ متغیری با عنوان Churn یا متغیری که مستقیماً وضعیت ریزش مشتری را مشخص کند وجود ندارد.

بنابراین، دیتاست فاقد برچسب آماده Churn است.

به عبارت دیگر، برای هیچ مشتری به‌صورت مستقیم مشخص نشده است که آیا مشتری فعال است یا ریزش کرده است.

#### ۲. آیا نبود برچسب Churn به معنی نبود اطلاعات رفتاری است؟

خیر.

اگرچه برچسب مستقیم Churn وجود ندارد، اما دیتاست اطلاعات تراکنش‌های مشتریان را در طول زمان ثبت کرده است.

این اطلاعات می‌توانند شواهد رفتاری لازم برای تعریف عملیاتی ریزش مشتری را فراهم کنند.

مهم‌ترین شواهد موجود عبارت‌اند از:

| اطلاعات | ستون مرتبط | کاربرد در تحلیل رفتار مشتری |
|---|---|---|
| شناسه مشتری | CustomerID | اتصال تراکنش‌های یک مشتری |
| تاریخ خرید | InvoiceDate | بررسی زمان خریدها |
| شماره فاکتور | InvoiceNo | شناسایی فاکتورها و فاکتورهای لغوشده |
| تعداد کالا | Quantity | بررسی مقدار کالاهای ثبت‌شده |
| قیمت واحد | UnitPrice | بررسی اطلاعات قیمت تراکنش |
| کد محصول | StockCode | بررسی تنوع محصولات خریداری‌شده |

این اطلاعات امکان بررسی رفتار خرید مشتریان را فراهم می‌کنند.

برای مثال، می‌توانیم بررسی کنیم:
- هر مشتری چند بار خرید کرده است؟
- آخرین خرید مشتری چه زمانی بوده است؟
- فاصله زمانی بین خریدهای مشتری چقدر بوده است؟
- آیا مشتری در یک بازه زمانی مشخص خرید داشته است؟
- آیا پس از یک دوره خرید، فعالیت مشتری متوقف شده است؟

بنابراین، اگرچه برچسب Churn در داده خام وجود ندارد، شواهد رفتاری لازم برای تعریف آن تا حد زیادی در دیتاست موجود است.

#### ۳. چرا باید Churn را به‌صورت عملیاتی تعریف کنیم؟

Churn مفهومی است که باید متناسب با نوع کسب‌وکار و رفتار مشتری تعریف شود.

برای مثال، اگر مشتری یک فروشگاه اینترنتی در طول یک ماه خرید نکند، لزوماً نمی‌توان گفت که برای همیشه فروشگاه را ترک کرده است.

ممکن است فاصله زمانی طبیعی بین خریدهای او بیشتر از یک ماه باشد.

بنابراین، برای جلوگیری از تعریف دلخواه و غیرقابل‌اندازه‌گیری، باید یک معیار زمانی مشخص برای ریزش در نظر بگیریم.

در پروژه ما، تعریف عملیاتی Churn بر اساس دو بازه زمانی انجام می‌شود:

**الف) Observation Window**

بازه 180 روزه برای مشاهده رفتار گذشته مشتری.

در این بازه بررسی می‌کنیم که مشتری چه رفتار خریدی داشته و آیا سابقه خرید معتبری دارد یا خیر.

**ب) Churn Window**

بازه 90 روزه پس از پایان Observation Window.

در این بازه بررسی می‌کنیم که آیا مشتری خرید معتبر دیگری انجام داده است یا خیر.

#### ۴. تعریف عملیاتی پیشنهادی Churn

با استفاده از این دو بازه زمانی، تعریف عملیاتی پروژه به شکل زیر خواهد بود:

- مشتری باید در بازه 180 روزه مشاهده، حداقل یک خرید معتبر داشته باشد.
- اگر مشتری در بازه 90 روزه بعد از آن هیچ خرید معتبری نداشته باشد، به‌عنوان Churn در نظر گرفته می‌شود.
- اگر مشتری در بازه 90 روزه بعدی حداقل یک خرید معتبر داشته باشد، به‌عنوان Non-Churn در نظر گرفته می‌شود.

به شکل ساده:

| وضعیت مشتری | Observation Window | Churn Window | برچسب |
|---|---|---|---|
| مشتری دارای خرید در گذشته و بدون خرید در آینده | دارای خرید | بدون خرید | Churn = 1 |
| مشتری دارای خرید در گذشته و دارای خرید در آینده | دارای خرید | دارای خرید | Churn = 0 |
| مشتری بدون خرید در Observation Window | بدون خرید | هر وضعیتی | خارج از جامعه هدف این تعریف |

نکته مهم این است که مشتریانی که در بازه مشاهده هیچ خرید معتبری ندارند، با این تعریف در گروه مشتریان قابل برچسب‌گذاری قرار نمی‌گیرند.

#### ۵. چرا Observation Window و Churn Window هر دو لازم هستند؟

اگر فقط تاریخچه خرید مشتری را بررسی کنیم، می‌توانیم میزان فعالیت گذشته او را بسنجیم؛ اما نمی‌توانیم با اطمینان مشخص کنیم که آیا واقعاً خریدهایش متوقف شده است یا خیر.

برای تعریف Churn باید رفتار آینده مشتری را نیز در یک بازه مشخص مشاهده کنیم.

بنابراین:

- Observation Window برای استخراج اطلاعات رفتاری گذشته استفاده می‌شود.
- Churn Window برای تعیین وضعیت بعدی مشتری و ساخت برچسب استفاده می‌شود.

این تفکیک زمانی باعث می‌شود اطلاعات گذشته و وضعیت آینده مشتری با یکدیگر اشتباه نشوند.

#### ۶. محدودیت‌های تعریف Churn در این دیتاست

با وجود اطلاعات رفتاری مناسب، چند محدودیت وجود دارد:

1. دیتاست ستون آماده Churn ندارد.
2. حدود 24.93 درصد رکوردها فاقد CustomerID هستند و نمی‌توان آن‌ها را با اطمینان به مشتری مشخصی نسبت داد.
3. بازه زمانی دیتاست محدود به حدود یک سال است؛ بنابراین فقط برای برخی تاریخ‌های مرجع امکان تشکیل کامل هر دو پنجره 180 و 90 روزه وجود دارد.
4. خریدهای لغوشده، مقادیر منفی یا صفر Quantity و قیمت‌های صفر یا منفی باید در تعریف خرید معتبر به‌دقت بررسی شوند.
5. نبود خرید در یک بازه 90 روزه لزوماً به معنی ترک دائمی فروشگاه نیست؛ بلکه در این پروژه یک تعریف عملیاتی و قابل‌اندازه‌گیری از Churn است.

# تکمیل ممیزی و تحلیل اکتشافی (Data Understanding – تکمیلی)

در این بخش، مواردی که در بررسی اولیه نیازمند عمق بیشتری بودند تحلیل می‌شوند. داده خام دست‌نخورده باقی می‌ماند؛ ماسک‌ها و ستون‌های کمکی فقط برای تحلیل ساخته می‌شوند. دسته‌بندی‌های این بخش، قواعد ممیزی اولیه هستند و پیش از Data Preparation باید از نظر کسب‌وکاری تأیید شوند.

**نکته روش‌شناختی:** مبلغ `Quantity × UnitPrice` در این دیتاست «ارزش خط فاکتور» است، نه لزوماً درآمد شناسایی‌شده یا وجه دریافتی. همچنین `CustomerID` گمشده را نمی‌توان به مشتری ساختگی واحدی نسبت داد.

</div>

```python
# Imports and non-destructive working view
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

analysis_df = df.copy()
analysis_df["InvoiceDateParsed"] = pd.to_datetime(
    analysis_df["InvoiceDate"], format="%m/%d/%y %H:%M", errors="coerce"
)
analysis_df["InvoiceNoText"] = analysis_df["InvoiceNo"].astype("string").str.strip()
analysis_df["StockCodeText"] = analysis_df["StockCode"].astype("string").str.strip().str.upper()
analysis_df["DescriptionText"] = analysis_df["Description"].astype("string").str.strip()
analysis_df["LineValue"] = analysis_df["Quantity"] * analysis_df["UnitPrice"]
analysis_df["IsCancelled"] = analysis_df["InvoiceNoText"].str.startswith("C", na=False)

print("Rows:", len(analysis_df))
print("Unparseable dates:", analysis_df["InvoiceDateParsed"].isna().sum())
print("pandas:", pd.__version__)

# 1) Non-merchandise / service-like StockCodes: inspect, do not blindly remove.
non_merchandise_codes = {
    "POST", "DOT", "M", "BANK CHARGES", "AMAZONFEE", "CRUK", "D"
}
non_merchandise_mask = analysis_df["StockCodeText"].isin(non_merchandise_codes)
non_merchandise_summary = (
    analysis_df.loc[non_merchandise_mask]
    .groupby(["StockCodeText", "DescriptionText"], dropna=False)
    .agg(
        Rows=("InvoiceNo", "size"),
        Invoices=("InvoiceNoText", "nunique"),
        Customers=("CustomerID", "nunique"),
        QuantitySum=("Quantity", "sum"),
        UnitPriceMin=("UnitPrice", "min"),
        UnitPriceMax=("UnitPrice", "max"),
        LineValueSum=("LineValue", "sum"),
    )
    .sort_values("Rows", ascending=False)
)
display(non_merchandise_summary)

# Inspect the highest unit prices and their transaction context.
display(
    analysis_df.nlargest(20, "UnitPrice")[[
        "InvoiceNoText", "StockCodeText", "DescriptionText", "Quantity",
        "UnitPrice", "LineValue", "CustomerID", "Country", "IsCancelled"
    ]]
)

# 2) Missing CustomerID: row, invoice, country, month, invoice type and value.
missing_id_mask = analysis_df["CustomerID"].isna()
analysis_df["Month"] = analysis_df["InvoiceDateParsed"].dt.to_period("M").astype("string")
analysis_df["InvoiceType"] = np.where(analysis_df["IsCancelled"], "Cancelled", "Non-cancelled")

missing_id_by_country = (
    analysis_df.groupby("Country", dropna=False)
    .agg(
        Rows=("CustomerID", "size"),
        MissingIDRows=("CustomerID", lambda s: s.isna().sum()),
        Invoices=("InvoiceNoText", "nunique"),
        MissingIDInvoices=("InvoiceNoText", lambda s: s[analysis_df.loc[s.index, "CustomerID"].isna()].nunique()),
        LineValue=("LineValue", "sum"),
    )
)
missing_id_by_country["MissingIDRowPct"] = (
    100 * missing_id_by_country["MissingIDRows"] / missing_id_by_country["Rows"]
)
display(missing_id_by_country.sort_values("MissingIDRows", ascending=False).head(20))

missing_id_by_month = (
    analysis_df.groupby("Month", dropna=False)
    .agg(
        Rows=("CustomerID", "size"),
        MissingIDRows=("CustomerID", lambda s: s.isna().sum()),
        LineValue=("LineValue", "sum"),
        MissingIDLineValue=("LineValue", lambda s: s[analysis_df.loc[s.index, "CustomerID"].isna()].sum()),
    )
)
missing_id_by_month["MissingIDRowPct"] = 100 * missing_id_by_month["MissingIDRows"] / missing_id_by_month["Rows"]
display(missing_id_by_month)

missing_id_by_type = (
    analysis_df.groupby("InvoiceType")
    .agg(
        Rows=("CustomerID", "size"),
        MissingIDRows=("CustomerID", lambda s: s.isna().sum()),
        Invoices=("InvoiceNoText", "nunique"),
        LineValue=("LineValue", "sum"),
    )
)
missing_id_by_type["MissingIDRowPct"] = 100 * missing_id_by_type["MissingIDRows"] / missing_id_by_type["Rows"]
display(missing_id_by_type)

missing_id_by_price = (
    analysis_df.assign(PriceGroup=np.select(
        [analysis_df["UnitPrice"].eq(0), analysis_df["UnitPrice"].lt(0)],
        ["Zero price", "Negative price"], default="Positive price"
    ))
    .groupby("PriceGroup")
    .agg(
        Rows=("CustomerID", "size"),
        MissingIDRows=("CustomerID", lambda s: s.isna().sum()),
        LineValue=("LineValue", "sum"),
    )
)
missing_id_by_price["MissingIDRowPct"] = 100 * missing_id_by_price["MissingIDRows"] / missing_id_by_price["Rows"]
display(missing_id_by_price)

invoice_level_missing = (
    analysis_df.groupby("InvoiceNoText", dropna=False)
    .agg(
        CustomerIDCount=("CustomerID", "nunique"),
        HasMissingCustomerID=("CustomerID", lambda s: s.isna().any()),
        AllRowsMissingCustomerID=("CustomerID", lambda s: s.isna().all()),
        InvoiceLineValue=("LineValue", "sum"),
        InvoiceType=("InvoiceType", "first"),
        Country=("Country", "first"),
    )
)
missing_invoice_summary = invoice_level_missing.loc[
    invoice_level_missing["AllRowsMissingCustomerID"]
].agg({
    "InvoiceLineValue": ["count", "sum", "mean", "median", "min", "max"]
})
display(missing_invoice_summary)

# 3) Customer-level cancellation / purchase contradiction.
customer_invoice_flags = (
    analysis_df.dropna(subset=["CustomerID"])
    .groupby("CustomerID")
    .agg(
        AnyCancelled=("IsCancelled", "any"),
        HasNonCancelledInvoice=("IsCancelled", lambda s: (~s).any()),
        CancelledInvoices=("InvoiceNoText", lambda s: s[analysis_df.loc[s.index, "IsCancelled"]].nunique()),
        AllInvoices=("InvoiceNoText", "nunique"),
    )
)
customer_invoice_flags["CancelledOnlyCustomer"] = (
    customer_invoice_flags["AnyCancelled"] & ~customer_invoice_flags["HasNonCancelledInvoice"]
)
print("Unique identified customers:", analysis_df["CustomerID"].nunique())
print("Customers with at least one non-cancelled invoice:", int(customer_invoice_flags["HasNonCancelledInvoice"].sum()))
print("Customers with cancelled invoices:", int(customer_invoice_flags["AnyCancelled"].sum()))
print("Customers with cancelled invoices only:", int(customer_invoice_flags["CancelledOnlyCustomer"].sum()))
display(customer_invoice_flags["CancelledOnlyCustomer"].value_counts().rename_axis("CancelledOnlyCustomer").to_frame("Customers"))

# 4) Match cancellation lines to prior positive lines for same customer and StockCode.
# This is a candidate matching analysis, not proof of exact order linkage.
line_cols = ["CustomerID", "StockCodeText", "InvoiceDateParsed", "Quantity", "InvoiceNoText", "UnitPrice"]
purchase_lines = analysis_df.loc[
    (~analysis_df["IsCancelled"]) & analysis_df["CustomerID"].notna()
    & analysis_df["Quantity"].gt(0) & analysis_df["UnitPrice"].gt(0), line_cols
].copy()
cancel_lines = analysis_df.loc[
    analysis_df["IsCancelled"] & analysis_df["CustomerID"].notna()
    & analysis_df["Quantity"].lt(0), line_cols
].copy()
purchase_lines = purchase_lines.sort_values("InvoiceDateParsed")
cancel_lines = cancel_lines.sort_values("InvoiceDateParsed")
# Rename the time columns explicitly so the source dates remain distinguishable.
purchase_lines = purchase_lines.rename(columns={
    "InvoiceDateParsed": "PurchaseDate",
    "InvoiceNoText": "PurchaseInvoiceNo",
    "Quantity": "PurchaseQuantity",
})
cancel_lines = cancel_lines.rename(columns={
    "InvoiceDateParsed": "CancelDate",
    "InvoiceNoText": "CancelInvoiceNo",
    "Quantity": "CancelQuantity",
})
matched_cancel_candidates = pd.merge_asof(
    cancel_lines.sort_values("CancelDate"),
    purchase_lines.sort_values("PurchaseDate"),
    by=["CustomerID", "StockCodeText"],
    left_on="CancelDate", right_on="PurchaseDate",
    direction="backward", allow_exact_matches=True,
    suffixes=("_cancel", "_purchase"),
)
matched_cancel_candidates["QuantityMatch"] = (
    matched_cancel_candidates["CancelQuantity"].abs()
    .eq(matched_cancel_candidates["PurchaseQuantity"])
)
matched_cancel_candidates["DaysToCancellation"] = (
    matched_cancel_candidates["CancelDate"] - matched_cancel_candidates["PurchaseDate"]
).dt.total_seconds() / 86400
matched_cancel_candidates["HasPriorCandidate"] = matched_cancel_candidates["PurchaseDate"].notna()
print("Cancellation lines with prior same-customer/same-product candidate:", int(matched_cancel_candidates["HasPriorCandidate"].sum()))
print("Of those, equal absolute quantity:", int((matched_cancel_candidates["HasPriorCandidate"] & matched_cancel_candidates["QuantityMatch"]).sum()))
display(matched_cancel_candidates[[
    "CancelInvoiceNo", "PurchaseInvoiceNo", "CustomerID", "StockCodeText",
    "CancelQuantity", "PurchaseQuantity", "QuantityMatch", "DaysToCancellation"
]].head(20))

# 5) Zero-price and non-C negative-quantity rows: inspect descriptions and IDs.
zero_price_rows = analysis_df[analysis_df["UnitPrice"].eq(0)]
negative_non_cancelled_zero_price = analysis_df[
    (~analysis_df["IsCancelled"]) & analysis_df["Quantity"].lt(0) & analysis_df["UnitPrice"].eq(0)
]
positive_zero_price = analysis_df[
    analysis_df["Quantity"].gt(0) & analysis_df["UnitPrice"].eq(0)
]
print("Zero-price rows:", len(zero_price_rows))
print("Negative quantity, non-cancelled, zero-price rows:", len(negative_non_cancelled_zero_price))
print("Positive quantity, zero-price rows:", len(positive_zero_price))
display(
    zero_price_rows.groupby(["StockCodeText", "DescriptionText"], dropna=False)
    .agg(Rows=("InvoiceNoText", "size"), Customers=("CustomerID", "nunique"),
         MissingCustomerID=("CustomerID", lambda s: s.isna().sum()),
         QuantitySum=("Quantity", "sum"))
    .sort_values("Rows", ascending=False).head(40)
)

# 6) Customer purchase cadence, eligible population and illustrative churn by reference date.
# Candidate valid merchandise purchases: non-cancelled, positive quantity/price,
# identified customer, and not one of the explicitly flagged service codes.
valid_merchandise_line_mask = (
    (~analysis_df["IsCancelled"])
    & analysis_df["InvoiceDateParsed"].notna()
    & analysis_df["Quantity"].gt(0)
    & analysis_df["UnitPrice"].gt(0)
    & ~analysis_df["StockCodeText"].isin(non_merchandise_codes)
)
valid_purchase_mask = valid_merchandise_line_mask & analysis_df["CustomerID"].notna()
valid_purchase_lines = analysis_df.loc[valid_purchase_mask].copy()
customer_purchase_dates = (
    valid_purchase_lines[["CustomerID", "InvoiceDateParsed"]]
    .drop_duplicates()
    .rename(columns={"InvoiceDateParsed": "PurchaseDate"})
    .sort_values(["CustomerID", "PurchaseDate"])
)
customer_purchase_dates["DaysBetweenPurchases"] = (
    customer_purchase_dates.groupby("CustomerID")["PurchaseDate"].diff().dt.total_seconds() / 86400
)
interpurchase_gaps = customer_purchase_dates["DaysBetweenPurchases"].dropna()
print("Customer purchase-gap observations:", len(interpurchase_gaps))
if not interpurchase_gaps.empty:
    display(interpurchase_gaps.describe(percentiles=[.25, .5, .75, .9, .95]).to_frame("DaysBetweenPurchases"))

# A reference date grid: month starts that allow full 180-day observation + 90-day future window.
first_date = analysis_df["InvoiceDateParsed"].min()
last_date = analysis_df["InvoiceDateParsed"].max()
reference_dates = pd.date_range(
    start=(first_date + pd.Timedelta(days=180)).normalize(),
    end=(last_date - pd.Timedelta(days=90)).normalize(),
    freq="MS",
)
valid_by_customer = valid_purchase_lines[["CustomerID", "InvoiceDateParsed"]].copy()
reference_results = []
for reference_date in reference_dates:
    observation_start = reference_date - pd.Timedelta(days=180)
    churn_end = reference_date + pd.Timedelta(days=90)
    observed = valid_by_customer[
        valid_by_customer["InvoiceDateParsed"].ge(observation_start)
        & valid_by_customer["InvoiceDateParsed"].lt(reference_date)
    ]
    future = valid_by_customer[
        valid_by_customer["InvoiceDateParsed"].ge(reference_date)
        & valid_by_customer["InvoiceDateParsed"].lt(churn_end)
    ]
    eligible_ids = pd.Index(observed["CustomerID"].unique())
    future_active_ids = pd.Index(future["CustomerID"].unique())
    churned_count = int((~eligible_ids.isin(future_active_ids)).sum())
    eligible_count = len(eligible_ids)
    reference_results.append({
        "ReferenceDate": reference_date,
        "EligibleCustomers": eligible_count,
        "FutureActiveCustomers": int(eligible_ids.isin(future_active_ids).sum()),
        "OperationalChurnCount": churned_count,
        "InitialChurnRatePct": (100 * churned_count / eligible_count) if eligible_count else np.nan,
    })
reference_date_summary = pd.DataFrame(reference_results)
display(reference_date_summary)

# Recency distribution at each candidate reference date, among eligible customers.
recency_tables = []
for reference_date in reference_dates:
    obs = valid_by_customer[
        valid_by_customer["InvoiceDateParsed"].lt(reference_date)
        & valid_by_customer["InvoiceDateParsed"].ge(reference_date - pd.Timedelta(days=180))
    ]
    last_purchase = obs.groupby("CustomerID")["InvoiceDateParsed"].max()
    recency = (reference_date - last_purchase).dt.total_seconds() / 86400
    recency_tables.append({
        "ReferenceDate": reference_date,
        "Customers": len(recency),
        "RecencyMedianDays": recency.median(),
        "RecencyP75Days": recency.quantile(.75),
        "RecencyP90Days": recency.quantile(.90),
    })
recency_by_reference = pd.DataFrame(recency_tables)
display(recency_by_reference)

# 7) Monthly metrics: active identified customers, invoices, line value.
monthly_metrics = (
    analysis_df.groupby("Month")
    .agg(
        TransactionRows=("InvoiceNoText", "size"),
        UniqueInvoices=("InvoiceNoText", "nunique"),
        ActiveIdentifiedCustomers=("CustomerID", "nunique"),
        GrossLineValue=("LineValue", "sum"),
    )
)
monthly_metrics.index = monthly_metrics.index.astype(str)
display(monthly_metrics)

weekday_counts = (
    analysis_df.dropna(subset=["InvoiceDateParsed"])
    .assign(Weekday=analysis_df["InvoiceDateParsed"].dt.day_name())
    .groupby("Weekday").size()
    .reindex(["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"], fill_value=0)
    .rename("Rows").to_frame()
)
display(weekday_counts)
print("Saturday rows:", int(weekday_counts.loc["Saturday", "Rows"]))

# 8) Mutually exclusive row audit. Priority rules prevent double counting.
# 'No CustomerID' is a separate final class only for otherwise valid merchandise lines.
record_type = pd.Series("Other / needs review", index=analysis_df.index, dtype="string")
record_type.loc[analysis_df["IsCancelled"]] = "Cancelled invoice line"
record_type.loc[(~analysis_df["IsCancelled"]) & analysis_df["StockCodeText"].isin(non_merchandise_codes)] = "Non-merchandise/service"
record_type.loc[(~analysis_df["IsCancelled"]) & ~analysis_df["StockCodeText"].isin(non_merchandise_codes)
                & (analysis_df["Quantity"].eq(0) | analysis_df["UnitPrice"].lt(0))] = "Adjustment / anomalous"
record_type.loc[(~analysis_df["IsCancelled"]) & ~analysis_df["StockCodeText"].isin(non_merchandise_codes)
                & analysis_df["Quantity"].gt(0) & analysis_df["UnitPrice"].eq(0)] = "Free / zero-price line"
record_type.loc[valid_merchandise_line_mask & analysis_df["CustomerID"].isna()] = "Valid-like line without CustomerID"
record_type.loc[valid_purchase_mask] = "Candidate valid merchandise purchase"
analysis_df["AuditRecordType"] = record_type
row_audit = (
    analysis_df["AuditRecordType"].value_counts(dropna=False)
    .rename_axis("RecordType").to_frame("Rows")
)
row_audit["Percentage"] = 100 * row_audit["Rows"] / len(analysis_df)
display(row_audit)
assert row_audit["Rows"].sum() == len(analysis_df), "Audit categories must cover each row exactly once."
assert analysis_df["AuditRecordType"].notna().all(), "Every row must have exactly one audit type."
print("Audit total:", int(row_audit["Rows"].sum()), "of", len(analysis_df))

# 9) Robust plots: 99th percentile zoom and signed log scale.
for column_name in ["Quantity", "UnitPrice"]:
    values = analysis_df[column_name].dropna()
    upper_limit = values.quantile(0.99)
    lower_limit = values.quantile(0.01)
    central_values = values.clip(lower=lower_limit, upper=upper_limit)
    plt.figure(figsize=(10, 4))
    plt.hist(central_values, bins=60)
    plt.title(f"{column_name} distribution (winsorized view: 1st–99th percentile)")
    plt.xlabel(column_name)
    plt.ylabel("Frequency")
    plt.tight_layout()
    plt.show()

    plt.figure(figsize=(10, 3))
    plt.boxplot(central_values, orientation="horizontal")
    plt.title(f"{column_name} boxplot (1st–99th percentile view)")
    plt.xlabel(column_name)
    plt.tight_layout()
    plt.show()

# 10) Country comparison: UK vs non-UK, and low-customer country warning.
country_view = analysis_df.copy()
country_view["CountryGroup"] = np.where(
    country_view["Country"].eq("United Kingdom"), "UK", "Non-UK"
)
uk_comparison = (
    country_view.groupby("CountryGroup")
    .agg(
        Rows=("InvoiceNoText", "size"),
        Invoices=("InvoiceNoText", "nunique"),
        IdentifiedCustomers=("CustomerID", "nunique"),
        MissingCustomerIDRows=("CustomerID", lambda s: s.isna().sum()),
        GrossLineValue=("LineValue", "sum"),
        MeanUnitPrice=("UnitPrice", "mean"),
    )
)
display(uk_comparison)
country_customer_quality = (
    analysis_df.groupby("Country", dropna=False)
    .agg(Rows=("Country", "size"), Invoices=("InvoiceNoText", "nunique"),
         IdentifiedCustomers=("CustomerID", "nunique"),
         MissingCustomerIDRows=("CustomerID", lambda s: s.isna().sum()))
    .sort_values("IdentifiedCustomers")
)
display(country_customer_quality.head(20))
print("Country labels containing 'unspecified' or 'unknown':")
print(country_customer_quality[
    country_customer_quality.index.to_series().astype(str).str.contains(
        "unspecified|unknown|other|n/a", case=False, regex=True
    )
])
```

## تفسیر و محدودیت‌های تحلیل تکمیلی

- فهرست کدهای غیرکالایی در این بخش، فهرست اولیه برای بازبینی است؛ کدهای مبهم باید با توضیحات و اسناد کسب‌وکار تأیید شوند.
- تطبیق لغو با خرید قبلی فقط «نامزد تطبیق» بر اساس مشتری، کالا و زمان است. به دلیل نبود شناسه سفارش اصلی/مرجوعی، این تطبیق قطعی نیست و یک خرید ممکن است چند ردیف مشابه داشته باشد.
- نرخ ریزش محاسبه‌شده برای تاریخ‌های مرجع، برآورد عملیاتی بر اساس تعریف ۱۸۰/۹۰ روزه و قواعد خرید معتبر همین بخش است؛ نرخ قطعی کسب‌وکار نیست. انتخاب تاریخ مرجع نهایی باید با هدف پروژه و پوشش زمانی هماهنگ شود.
- `GrossLineValue` جمع خام Quantity×UnitPrice است و ممکن است هزینه‌ها، تعدیلات، لغوها و ردیف‌های غیرکالایی را دربر بگیرد؛ آن را بدون پالایش و تعریف مالی، درآمد خالص تلقی نکنید.
- جدول ممیزی به‌صورت انحصاری ساخته شده و مجموع دسته‌ها باید دقیقاً با تعداد کل رکوردها برابر باشد. این دسته‌بندی برای ممیزی است، نه تصمیم حذف قطعی.
- تفاوت UK و Non-UK در این مرحله توصیفی است؛ برای ادعای تفاوت آماری یا پیش‌بینی‌پذیری باید نمونه کافی، تعریف مشتری و آزمون/مدل مناسب بررسی شود.
